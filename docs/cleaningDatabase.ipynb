```python
code = """

{{PEthing|Kal-El (New Earth)|Superman}}
{{DC Database:Character Template 
| PageTitle               = Kal-El (Prime Earth)
| Image                   = <!--Do not change the main image without first suggesting it on the talk page. Changes without consultation will be reverted. Repeat offenders will be blocked. -->Action Comics Vol 1 1000 Textless.jpg <!--Do not change the main image without first suggesting it on the talk page. Changes without consultation will be reverted. Repeat offenders will be blocked. -->
| RealName                = [[Kal-El]] (Birth Name)<br/>[[Clark Kent|Clark Joseph Kent]] (Legal Name)
| MainAlias               = Superman
| Aliases                 = Aksel;<br>Chaz Donen;<br>God of Strength;<br>Jon Clark;<br>[[Parallax]];<br>[[Yellow Lantern]];<br> [[Superman Blue]];<br>[[Superboy]];<br>[[Doctor Fate]]<ref>{{c|Superman Vol 5 24}}</ref>;<br>Bahle'na Gahl<ref>{{c|Action Comics Vol 1 1040}}</ref>;<br>[[Bizarro No. 1|Bizarro Number One]]<ref>{{c|Action Comics Vol 1 1061}}</ref>;<br>Klar-Ken
| Identity                = Secret<!-- Action Comics Vol 1 1050 -->
| Alignment               = Good
| Affiliation             = [[Justice League (Prime Earth)|Justice League Unlimited]],<br> [[LexCorp|SuperCorp]],<br> [[Daily Planet]],<br> [[Superman Family]],<br> [[Authority (Prime Earth)|The Authority]];<br> Formerly [[Justice League: Team Mystery]],<br> [[Sinestro Corps (Prime Earth)|Sinestro Corps]],<br> [[Legion of Super-Heroes (Pre-Zero Hour)|Legion of Super-Heroes]]
| Relatives               = [[Lois Lane (Prime Earth)|Lois Lane-Kent]] (wife)<br>[[Jor-El (Prime Earth)|Jor-El]] (father, deceased)<br>[[Lara Lor-Van (Prime Earth)|Lara Lor-Van]] (mother, deceased)<br>[[Nim-El (Prime Earth)|Nim-El]] (uncle, deceased)<br>[[Zor-El (Prime Earth)|Zor-El]] (uncle, deceased)<br>[[Alura In-Ze (Prime Earth)|Alura In-Ze]] (aunt, deceased)<br>[[Kara Zor-El (Prime Earth)|Kara Zor-El]] (cousin)<br>[[Jonathan Kent (Prime Earth)|Jonathan Kent]] (adoptive father)<br>[[Martha Clark (Prime Earth)|Martha Kent]] (adoptive mother)<br>[[Jonathan Samuel Kent (Prime Earth)|Jon Kent]] (son)<br>[[Lar Gand (Prime Earth)|Lar Gand]] (adoptive brother)<br>[[Otho-Ra (Prime Earth)|Otho-Ra]] (adopted daughter)<br>[[Osul-Ra (Prime Earth)|Osul-Ra]] (adopted son)<br>[[Mon-El (Post-Rebirth)|Mon-El]] (sixth great-grandson)<br>[[Samuel Lane (Prime Earth)|Sam Lane]] (father in-law, deceased)<br>[[Elenor Lane (Prime Earth)|Elenor Lane]] (mother-in-law, deceased)<br>[[Leonardo Lane (Prime Earth)|Leo Lane]] (brother-in-law)<br>[[Lucy Lane (Prime Earth)|Lucy Lane]] (sister-in-law)<br><small> '''Clones''': </small>[[Kon-El (Prime Earth)|Conner Kent]] (partial clone)<br>[[Subject B-0 (Prime Earth)|Bizarro]] (clone, deceased)<br>[[Bizarro II (Prime Earth)|Bizarro II]] (clone)<br>[[Super-Man Zero (Prime Earth)|Emperor Super-Man]] (partial clone)<br>[[Milkman Man (Prime Earth)|Milkman Man]] (partial clone, deceased)<br>[[Nuclear Man (Prime Earth)|Nuclear Man]] (clone, deceased)<br>see [[House of El]] and [[Kent Family]] for extended family line
| Universe                = Prime Earth
| BaseOfOperations        = [[Metropolis]], [[Fortress of Solitude]]; Formerly [[Justice League Watchtower]], [[Smallville]], Hamilton County

| Gender                  = Male
| Height                  = 6'3"<!-- Batman Character Encyclopedia -->
| Weight                  = 235 lbs <!-- Batman Character Encyclopedia -->
| Eyes                    = Blue
| Hair                    = Black
| UnusualFeatures         = 

| AlienRace               = Kryptonian
| Citizenship             = American
| MaritalStatus           = Married
| Occupation              = [[Superhero]], [[Reporter]]; Former [[Farmer]]

| Creators                = Jerry Siegel; Joe Shuster
| First                   = Flashpoint Vol 2 5

| Quotation               = His whole [[Krypton|planet]] was destroyed. He's the last of a '''holocaust'''. He grew up in the dirt. Finding out slowly how different he was. A stranger discovering every day how strange he was. He has the power to tear [[Earth|the world]] apart. And he could. With a pinkie. It's not his world. [[Humans|We]]'re not his people. We should be ants to him. Imagine that. Always being on the outside. The pain that would come from always being on the outside. And yet, he took that pain and became the symbol of hope.
| Speaker                 = [[Bruce Wayne (Prime Earth)|Batman]]
| QuoteSource             = Batman Vol 3 36

| Overview                = '''Superman''' (real name '''Clark Kent''', born '''Kal-El''') is one of the last children of [[Krypton]], sent as the dying planet's last hope to [[Earth]], where he grew to become its kind, noble protector. Using his flight, enhanced strength, heat vision, and various other powers, he protects the planet and the universe from powerful threats, all while inspiring hope into others. He is also a founding member of the [[Justice League (Prime Earth)|Justice League]], and the father of [[Jonathan Samuel Kent (Prime Earth)|Jon Kent]], the second Superman.
| HistoryText             = 
===Origins===
<div class="mw-collapsible mw-collapsed" style="width:100%">
====New 52 Origin====
<div class="mw-collapsible-content">
{{Image|Action_Comics_Vol_2_5_Textless.jpg|Baby Kal-El's escape from a dying planet|left|150px}}
Kal-El was born on Krypton to [[Jor-El (Prime Earth)|Jor-El]] and [[Lara Lor-Van (Prime Earth)|Lara]]. After Jor-El discovered that the planet was doomed to destruction, he attempted to escape with his family into the [[Phantom Zone]] but was repelled by the criminals who had already been imprisoned there. With no other choice, he and Lara resigned themselves to their fates, placing their son in a prototype evacuation [[Kryptonian Rocket|rocket]], activating its Brainiac intelligence, and sending it to a planet with a yellow sun: Earth.<ref name=AC-2-5>{{C|Action Comics Vol 2 5}}</ref>

The rocket landed in [[Smallville, Kansas]] where the Kryptonian baby was discovered by [[Jonathan Kent (Prime Earth)|Jonathan]] and [[Martha Clark (Prime Earth)|Martha Kent]], who had just learned that they could not have children of their own. Seeing this alien orphan as their chance, they gathered him up and took him home, leaving a stillborn calf in his place in the hopes of throwing the government off their trail.<ref name=AC-2-5/> They raised him as Clark Kent teaching him their values. Eventually, Jonathan revealed Clark's true parentage to him, but left it to him to decide what it would mean to him.

On the night of his prom, however, because Clark had no superhero role models to make him feel comfortable becoming a hero, Clark was unable to save his adopted parents from dying in a car accident.<ref name=SM-3-20>{{c|Superman Vol 3 20}}</ref> After their funerals, Clark sold [[Kent Farm]] to a neighbour and moved to Metropolis to become a journalist.<ref>{{C|Action Comics Vol 2 6}} (backup)</ref>

{{main|Superman and the Men of Steel}}
{{Image|Action_Comics_Vol_2_4_Textless_Variant.jpg|Superman debuts in Metropolis|right|150px}}
Upon moving to Metropolis, Clark took a job at the [[Daily Star]] and prepared himself to act as a vigilante at the age of 22,<ref name=BMSM3>{{c|Batman/Superman Vol 1 3}}</ref> by having several shirts printed with the [[House of El|S-Symbol]] that had come to Earth with him emblazoned on them. His attempt to rescue a young boy caused him to reveal himself before he expected to, and the city came to know of the superhuman man.<ref name="AC0">{{C|Action Comics Vol 2 0}}</ref>

Within six months, Clark had fully revealed himself and was targeting corrupt businessman [[Glen Glenmorgan (Prime Earth)|Glen Glenmorgan]] when [[Samuel Lane (Prime Earth)|General Sam Lane]] and [[Alexander Luthor (Prime Earth)|Lex Luthor]] took an interest in him. They staged several emergencies to trap him, finally succeeding with a train crash that endangered the General's own daughter [[Lois Lane (Prime Earth)|Lois]].<ref name="AC1">{{C|Action Comics Vol 2 1}}</ref> In collaboration with the military, Luthor tortured Superman while his [[Kryptonian Rocket]] was studied. Upon his escape, Superman sealed his birth rocket away for protection but was too late to prevent its [[Kryptonian]] signal from alerting the [[Vril Dox (Prime Earth)|Collector of Worlds]] to its location - and unaware that Luthor was complicit in making that happen.<ref>{{C|Action Comics Vol 2 2}}</ref>

After the district of [[New Troy]] had been secretly shrunken down and abducted by the Collector of Worlds, Superman made a giant leap from Metropolis into outer space, catching hold of the retreating spaceship. The Collector tried to force Superman to choose between the lives of those in the city of [[Kandor]] and of New Troy, but he refused to choose, instead donning a Kryptonian suit of armor from one of the bottles and promising to protect ''both''.<ref>{{C|Action Comics Vol 2 7}}</ref> Defeating the Collector and returning New Troy to its original place, Superman then came out officially to the people of Metropolis as an alien and promised to remain on the side of good.<ref name=AC8>{{C|Action Comics Vol 2 8}}</ref>
</div>
''Following the cosmic upheavals of the Rebirth period, this character's history was revised.''

====Current Origin====
{{Image|Superman_Reborn_002.jpg|Superman's new, composite origin|200px|left}}

Kal-El, the Last Son of Krypton, was born to scientists [[Jor-El (Prime Earth)|Jor-El]] and [[Lara Lor-Van (Prime Earth)|Lara Lor-Van]] shortly before the destruction of their doomed world.  Aware of their planet's impending fate, they conducted extensive (but remote) research on habitable worlds within range and constructed a [[Kryptonian Rocket|rocket]] so that their son might survive.<ref>{{C|World of Krypton Vol 3 6}}</ref> As an infant, he was introduced to his paternal cousin [[Kara Zor-El (Prime Earth)|Kara Zor-El]].  In the end, they chose [[Earth]], and managed to launch their son into space mere moments before the destruction of Krypton. Kal's rocket landed in [[Smallville, Kansas]], where he was discovered by [[Jonathan Kent (Prime Earth)|Jonathan]] and [[Martha Clark (Prime Earth)|Martha Kent]]. He was given the name "Clark Kent" and adopted as their son.

=====Childhood=====
When Clark began his schooling, Martha and Jonothan were at first concerned that he wouldn't fit in with his classmates, a worry that the young Clark shared.  Asking his father for advice, Jonathan simply tells Clark to tell jokes and if he gets people to laugh, they'll be his friends. Clark's first day isn't as difficult as he expects and he ends up making many friends at school, until he notices a girl, Samantha, sitting alone in the playground; ostracized for being different.

Clark decided that he wants to befriend the girl, but feared that if he did, he'd lose all the friends he'd made up to that point.  Consulting with his parents, they advise him to befriend the girl regardless of his classmates' opinions.  He followed through, and his example inspired others in his class to befriend Samantha.<ref>{{C|Superman Red and Blue Vol 1 1}}</ref>

=====Martha's Cancer=====

During Clark's childhood, Martha was diagnosed with cancer, and underwent chemotherapy for the affliction; Clark was not told until well into her treatment, when Clark struggled with a new student attending the school, [[Caleb Withers (Prime Earth)|Caleb Withers]]. Caleb was something of a bully, with misplaced aggression spurred on by his abusive father.  Clark struggled with the urge not to respond to Caleb's behavior in kind, something his mother talked him out of, telling him that it was better to help people find the good within themselves rather than correct their behavior by force.

As they were leaving, Caleb kicked a soccer ball into Martha's head, knocking her down and inadvertently removing her wig, revealing to Clark her hair loss.  Later that evening, he snuck over to Caleb's house, with the intent to destroy Caleb's bicycle in retribution, only to discover Caleb's drunk father abusing him.  As Caleb's father attempted to drive off, he nearly drove into oncoming traffic, with Clark saving his life.  He then brought the unconscious man to the local clinic, leaving him there anonymously.  Afterward, Clark attempted to cut off his own hair in solidarity with his mother, only to discover that his invulnerability made that impossible at the time. Later, Clark was able to reach Caleb, that he wasn't his father, and that he didn't have to be him.

Caleb eventually apologized to Martha, revealing that he'd cut his own hair in solidarity with Martha as part of his apology.  He left to live with his aunt in Metropolis, and one day would grow to run a community center for at-risk youth. <ref>{{c|Action Comics 2022 Annual Vol 1 1}}</ref>

=====Childhood Friends=====
Growing up in Smallville, Clark found himself becoming best friends with [[Lana Lang (Prime Earth)|Lana Lang]] and [[Peter Ross (Prime Earth)|Pete Ross]].  The trio were inseparable and did everything together, including hiding from the young [[Lex Luthor (Prime Earth)|Lex Luthor]].  The first time Clark flew was on instinct; a tornado hit Smallville, and he acted without thinking in order to save Lana Lang's life.<ref>{{C|Action Comics Vol 1 977}}</ref>

=====Krypto the Super-Dog=====
{{Image|Action Comics Vol 2 5 Textless Variant.jpg|Clark Kent and [[Krypto (Prime Earth)|Krypto]]|left|150px}}
One evening, during a family dinner, Clark's enhanced senses noticed an object crashing from space a few miles away.  He immediately set off to find it with super-speed, with his parents warning him not to touch it until they arrived.  Later, all three converged on the object; a Kryptonian space-pod lodged into a dry creek bed.  Clark immediately recognized it as resembling the pod he arrived in, and, reaching out to it, found that it recognized him in turn.  It opened, revealing a small dog.  The dog playfully jumped on Clark and began to display the same Kryptonian power-set that Clark possessed.  

Clark quickly sped off to play with the dog, easily bonding with it.  As he did, a holographic recording of his birth father, Jor-El, activates, revealing to his parents that the dog was a test subject for a prototype of the pod that eventually would bring Clark to Earth.  The recording also revealed the dog's name, given to him by the infant Kal-El: [[Krypto (Prime Earth)|Krypto]].  Approving of how happy Krypto made Clark, Jonathan and Martha allowed him to keep and raise the dog.<ref>{{C|Superman Red and Blue Vol 1 5}}</ref>

=====Teenage Years=====
One particular autumn during Clark's teenage years, the Kent farm suffered a bad harvest; in spite of this, Jonathan Kent decided to share what little they could spare with their neighbors, who had endured worse. The entire community helped each other out, surprising Clark, who had heard that some people disliked each other, and Jonathan explained that people ultimately want to be good, they just need an example to bring out the best side of themselves. Jonathan tells Clark to be that example.<ref>{{C|Superman Red and Blue Vol 1 4}}</ref>

At some point, Clark joined the [[Legion of Super-Heroes (Pre-Zero Hour)|Legion of Super-Heroes]] and fought alongside them in battles against the likes of [[Mordru (Prime Earth)|Mordru]] and [[Darkseid (Prime Earth)|Darkseid]].<ref>{{C|Justice Society of America Vol 4 11|Justice Society of America #11}}</ref><ref>{{C|DC All In Special Vol 1 1|DC All In Special #1}}</ref>

In a [[The New 52|previous iteration of history]], on the night of his senior prom, because his adoptive parents discouraged Clark from becoming a superhero for fear what it could do to him, he failed to save Jonathan and Martha from dying in a car accident.  This was not a natural occurence; it was arranged by the fifth-dimensional demon [[Vyndktvx (Prime Earth)|Vyndktvx]] as part of a multi-pronged attack on Clark at many points across the timeline.

However, when the [[Justice Society of America (Prime Earth)|Justice Society of America]] was restored to its proper place in the timeline, they set a good example for superheroes to come in the future. With role models to look up to and no reason to stay in hiding, on the night of his senior prom, Clark saved his adoptive parents from dying in a car accident.<ref name="DC12">{{C|Doomsday Clock Vol 1 12}}</ref>

=====Debut as Superman=====
{{main|Superman and the Men of Steel}}
{{Image|Action_Comics_Vol_2_4_Textless_Variant.jpg|Superman debuts in Metropolis|right|150px}}
Upon moving to Metropolis, Clark took a job at the [[Daily Star]] and prepared himself to act as a vigilante at the age of 22,<ref name=BMSM3>{{c|Batman/Superman Vol 1 3}}</ref> by having several shirts printed with the [[House of El|S-Symbol]] that had come to Earth with him emblazoned on them. His attempt to rescue a young boy caused him to reveal himself before he expected to, and the city came to know of the superhuman man.<ref name="AC0">{{C|Action Comics Vol 2 0}}</ref>

Superman and Batman met for the first time while Clark was investigating a series of murders in [[Wayne Enterprises]]. The two new heroes were unexpectedly whisked to a [[Earth 2|parallel reality]] by [[Kaiyo (Prime Earth)|Kaiyo the Trickster]] where they met [[Kal-El (Earth 2)|older]] and [[Bruce Wayne (Earth 2)|more experienced]] versions of themselves. When they were returned back to their Earth however, Kaiyo wiped their memories of the whole interaction.<ref>{{C|Batman/Superman Vol 1 1}}</ref>

During the [[Batman: Zero Year|Zero Year]], Clark attempted to personally disperse Hurricane Rene as it bore down on the coastline, still unsure of the upper limits of his powers.  His attempt to divert the hurricane was unsuccessful, but he did succeed in rescuing the tanker Jean-Marie, which had at the time been carrying his childhood friend, Lana Lang as its engineer.<ref>{{C|Action Comics Vol 2 25}}</ref>

Within six months, Clark had fully revealed himself and was targeting corrupt businessman [[Glen Glenmorgan (Prime Earth)|Glen Glenmorgan]] when [[Samuel Lane (Prime Earth)|General Sam Lane]] and [[Alexander Luthor (Prime Earth)|Lex Luthor]] took an interest in him. They staged several emergencies to trap him, finally succeeding with a train crash that endangered the General's own daughter [[Lois Lane (Prime Earth)|Lois]].<ref name="AC1">{{C|Action Comics Vol 2 1}}</ref> In collaboration with the military, Luthor tortured Superman while his [[Kryptonian Rocket]] was studied. Upon his escape, Superman sealed his birth rocket away for protection but was too late to prevent its [[Kryptonian]] signal from alerting the [[Vril Dox (Prime Earth)|Collector of Worlds]] to its location - and unaware that Luthor was complicit in making that happen.<ref name=AC2>{{C|Action Comics Vol 2 2}}</ref>

After the district of [[New Troy]] had been secretly shrunken down and abducted by the Collector of Worlds, Superman made a giant leap from Metropolis into outer space, catching hold of the retreating spaceship. The Collector tried to force Superman to choose between the lives of those in the city of [[Kandor]] and of New Troy, but he refused to choose, instead donning a Kryptonian suit of armor from one of the bottles and promising to protect ''both''.<ref>{{C|Action Comics Vol 2 7}}</ref> Defeating the Collector and returning New Troy to its original place, Superman then came out officially to the people of Metropolis as an alien and promised to remain on the side of good.<ref>{{C|Action Comics Vol 2 8}}</ref>  

===Early Career===
Reluctantly, after encouragement from his boss at the Daily Star, he accepted a job as a reporter at the [[Daily Planet]] where he was paired with [[Lois Lane (Prime Earth)|Lois Lane]].  On his first day, Clark saved Lois Lane's life during a helicopter crash, in the process debuting a new costume bearing red trunks and a low neckline, but clearly inspired by the Kryptonian armor Clark recovered from Brainiac's ship. Among the first villains he ever faced was the [[Ultra-Humanite II (Prime Earth)|Ultra-Humanite]], who developed a lifelong enmity towards him.<ref name=SatA-V1-4>{{C|Superman and the Authority Vol 1 4}}</ref>  He also quickly made an enemy of his childhood bully, now grown, Lex Luthor.

Early on in his career, before the founding of the [[Justice League]], Clark was displaced into the early 1960s during an unknown incident.  While there, he operated in secret- though not secret enough that he wasn't known to [[John F. Kennedy (Prime Earth)|the President of the United States]].  He met with the President, who requested that he, and others like him, come forward and use their collective abilities to make the world a better place.  He offered to protect the president in Dallas; Kennedy declined, insisting that Superman assist with a manned flight to the moon.  Shortly afterwards, the president was assassinated while riding in his motorcade in Dallas.  Eventually Superman was able to return to his own time; when he returned, he did his best to fufill his promise to the doomed president.<ref>{{C|Superman and the Authority Vol 1 1}}</ref>

====Forming the Justice League====
Clark was among the founding members of the [[Justice League (Prime Earth)|Justice League]]. While the team's actual first meeting is debated, prime amongst the contenders was when the heroes who would eventually comprise the league fought off an attack by the alien [[Appellaxians]], which resulted in them nearly being turned into trees before emerging victorious.  The heroes, inspired by their victory, formed a team not unlike the historical [[Justice Society of America (Prime Earth)|Justice Society of America]] or its lesser-known successor, the [[Justice Experience]], christening it the Justice League.  One of their earliest foes was [[Starro (Prime Earth)|Starro the Conquerer]].<ref>{{C|FCBD Dark Crisis Special Edition Vol 1 0}}</ref>

''In original New 52 Continuity, the Justice League formed following an invasion of Earth by [[Darkseid (Prime Earth)|Darkseid]].  However, following subsequent changes to the timeline, this was supplanted with an adventure resembling the team's [[Earth-1]] and [[New Earth]] origin.  While Darkseid's invasion may have still occurred, it seemingly did not occur in the way originally depicted.''

====Mongul and the Black Mercy====
At some point, he came into conflict with [[Mongul (Prime Earth)|Mongul MDCCXC]], an alien tyrant who ruled the vast [[Warworld]].  Later, this enemy would infect him with the [[Black Mercy]], an alien flower capable of inducing intense hallucinatory false realities.  While imprisoned within the Black Mercy's tendrils, Clark saw a version of Krypton that never exploded. In this world, Kal-El had a life on Krypton- a wife and a son.  Though he was eventually freed, the experience affected Clark deeply.  He ruminated on it for many years, and in the end, it inspired him to one day make a change and expand his circle of trust.<ref>{{c|Justice League Vol 4 52}}</ref>

===Further Adventures===
====World's Finest====
Over the course of their careers, Superman and Batman and became close allies, despite some initial distrust and the stark difference between their methods and personalities. They would frequently operate as partners, a pairing referred to by some as 'The World's Finest'.  At some point, they fought the tyrant Mongul MDCCXC together.

=====The Devil Nezha=====
One such outing was prompted by the unexpected team-up of the Gotham-based villain [[Pamela Isley (Prime Earth)|Poison Ivy]] and the cyborg [[John Corben (Prime Earth)|Metallo]]  under the influence of [[Nezha (Prime Earth)|the Devil Nezha]].  During the encounter, Metallo injected Superman with a sample of [[Red Kryptonite]], causing Superman to undergo a series of rapid and horrifying transformations. Batman contacted the Doom Patrol, who rapidly arrived on the scene.  Luckily, [[Lawrence Trainor (Prime Earth)|Negative Man]] was able to incapacitate Superman.<ref>{{C|Batman/Superman: World's Finest Vol 1 1}}</ref> The group was able to purge the Red Kryptonite from Superman; [[Niles Caulder (Prime Earth)|Niles Caulder]] then identified the insignia on the vial that had been used to inject Superman as belonging to the Devil Nezha (having recently recoverd Nezha's sword from [[General Zahl (Prime Earth)|General Zahl]] themselves, and recounted the legend of Nezha's creation.

Simultaneously, several villains attacked other heroes, including The Flash, Wonder Woman, Green Lantern, and the new hero [[Shazam (Billy Batson)|Captain Marvel]].  Splitting up, Robin teamed with [[Kara Zor-El (Prime Earth)|Supergirl]], and the pair traveled  in time in order to contact the [[House of Ji]], who had defeated and imprisoned Nezha in antiquity.  Superman and Batman traveled to Philadelphia in order to rescue Billy Batson from [[Felix Faust (Prime Earth)|Felix Faust]]<ref>{{C|Batman/Superman: World's Finest Vol 1 2}}</ref> Despite some struggle, they were able to defeat Faust and rescue Billy before moving on to aid other heroes.

The pair then went to rescue Wonder Woman and the Flash, who under attack by [[Albert Desmond (Prime Earth)|Doctor Alchemy]] and [[Samuel Scudder (Prime Earth)|Mirror Master]] respectively. They defeated both villains by tricking Doctor Alchemy into attacking Mirror Master's Mirror, only to be captured by [[Green Lantern (Hal Jordan)|Green Lantern]], who had fallen under Nezha's sway.<ref>{{C|Batman/Superman: World's Finest Vol 1 3}}</ref>  However, the pair was able to escape by overriding Hal Jordan's compromised willpower.  The ring, seeing them as one individual and amplified by the ambient magic produced by Nezha, fused Batman and Superman into a single composite being.<ref>{{C|Batman/Superman: World's Finest Vol 1 4}}</ref>  Nezha did not relent, though, and sent the heroes [[Dinah Drake (Prime Earth)|Black Canary]], [[Ronald Raymond (Prime Earth)|Firestorm]], Martian Manhunter, [[Flash (Wally West)|Kid Flash]], [[Donna Troy (Prime Earth)|Wonder Girl]], and [[Red Tornado (Prime Earth)|Red Tornado]] after them. However, the fused pair was able to force Nezha towards his far-away tomb with the assistance of the returned Supergirl. Realizing that Nezha was weak to modern technologies, the group managed to imprison Nezha within his tomb once again with the help of a [[Phantom Zone Projector]].<ref>{{C|Batman/Superman: World's Finest Vol 1 5}}</ref>

Afterwards, Superman and Batman travelled back in time to the year 1892 to rescue Robin, who had inadverdantly become lost in the timestream during his trip to learn from the House of Ji.  There, they participated in a circus show undercover, and helped Robin solve an a murder before the trio returned to the present.<ref>{{C|Batman/Superman: World's Finest Vol 1 6}}</ref>

=====The Boy Thunder=====
Some time later, a rift opened up in the sky, and through it, [[Magog (Earth-22)|David Sikela]] arrived, the last survivor from an unknown Earth elsewhere in them multiverse. They quickly realized the unique properties of David's biology gave him powers under Earth-0's sun, and brought him to Kandor to be examined by the Kandorian scientist [[Kim-Da (Prime Earth)|Kim-Da]].  While there, Kim-Da fabricated David a costume inspired by Hel-Oz, an ancient Kryptonian hero. After a brief but traumatic excursion to Gotham, Robin took David to the secret base of the [[Teen Titans (Prime Earth)|Teen Titans]], where he introduced the teen to his friends and partners.<ref>{{C|Batman/Superman: World's Finest Vol 1 7}}</ref>

For a time, Boy Thunder worked as Superman's sidekick; eventually, however, it would not last.  After several adventures in which he helped subdue an earthquake, battle Doctor Phosphorus, fight Sinestro, and save the city from Atomic Skull, things rapidly fell apart as David was directly targeted.  The [[Key (Prime Earth)|Key]] attacked Gotham, using a psychotropic compound to effectively lock all of its inhabitants where they were when the attack began, including Robin and Batman, who were stuck in the Batcave.  Despite an anxiety attack, David was able to help, rescuing trapped miners.  Luckily, based on Batman's notes, [[Theodore Kord (Prime Earth)|Blue Beetle]] was able to disperse an antidote to the chemical.  Unbeknownst to them, the Key was working with the Joker, who had larger plans.<ref>{{C|Batman/Superman: World's Finest Vol 1 10}}</ref>

David continued to study alongside Superman and Batman, struggling with anger issues and the trauma of losing his parents and world, later admitting to Superman that he believed himself to be responsible for his parents' death.  Their heart-to-heart was interrupted by the next stage of the Joker and the Key's plan- using [[Angle Man's Angler]] to open a door directly from the bottom of the Gotham River into the city.  Tragically, the pair used this distraction as an opportunity to kidnap David.  They tortured David as Superman and his allies desperately searched for their lost friend.  Eventually, they found him, and in the process of freeing him, David nearly killed the Joker.  In the aftermath, Superman briefly considered using Kryptonian technology designed by Kim-Da to depower David, but decided against it.

However, the Key was not yet done with David.  He summoned a motley crew of villains into the Fortress of Solitude, including Bag O' Bones, Acid Master, Doctor Phoenix, the Abominable Snowman, Ferlin Nxly, and Zebra Man.  In the chaos, he activated the Bleedship that brought David to Earth-0, seeing it as the key to the multiverse.  Resultingly, David was pulled out of Earth-0, sent to an unknown world in the Multiverse- but not before a tearful goodbye with Superman.<ref>{{C|Batman/Superman: World's Finest Vol 1 11}}</ref>

=====Elementary=====
<!-- Arc 3 of World's Finest -->
{{Incomplete}}

====The Death and Life of Superman ====
Over the years, Superman had many adventures.  At some point, some of Superman's DNA was collected by the [[Agenda]], [[Project Cadmus (Prime Earth)|Project Cadmus]] and [[Alexander Luthor (Prime Earth)|Lex Luthor]], resulting in multiple clones, including [[Kon-El (Prime Earth)|Superboy]] and [[Subject B-0 (Prime Earth)|Bizarro]]. After rescuing a group of children from a bus accident, a chance encounter with a nurse led Superman to dedicate a decent portion of his time to acting as a medical courier for organ donations. During one of his adventures, he was sent to a 'pocket earth', where he experienced a traumatic event that resulted in him leaving Earth for a time. While in space, Superman was forced to battle as a gladiator in one of Mongul's lesser Warworlds, and his refusal to kill his opponents led to a revolt. At some point later, Clark led a group of superheroes to defeat a planetary attack by Brainiac.<ref>{{C|The Death of Superman 30th Anniversary Special Vol 1 1}}</ref>.

{{Image|The Man of Steel Vol 2 5 Textless.jpg|Superman's death|left|150px}}
Over time, Clark and Lois fell in love, and he revealed his secret identity to her. Not long after, Superman was seemingly [[Death of Superman|killed by the alien monster Doomsday]]. In the wake of his death, four very different contenders arose to his mantle: The Cyborg [[Henry Henshaw (Prime Earth)|Hank Henshaw]], the erstwhile [[John Henry Irons (Prime Earth)|Steel]], the Cadmus-created clone [[Kon-El (Prime Earth)|Superboy]] and the android known as [[Eradicator II (Prime Earth)|the Eradicator.]]<ref name="YJ-V3-15>{{c|Young Justice Vol 3 15}}</ref> Eventually, the Cyborg Superman revealed his true colors, participating in the destruction of Coast City. Superman was soon restored to life; Steel and Superboy became lasting allies for the Man of Steel.<ref>{{C|Action Comics Vol 1 978}}</ref>.

====Marriage and Fatherhood====
Afterwards, Clark and Lois married. At one point, Superman's powers malfunctioned, and he temporarily became [[Superman Blue|Electric Blue Superman]] before returning to normal.<ref name=AC978>{{C|Action Comics Vol 1 978}}</ref>

Shortly after their marriage, Lois became pregnant with a child, and Clark delivered his son [[Jonathan Samuel Kent (Prime Earth)|Jonathan Samuel Kent]] in the [[Fortress of Solitude]] during an attempted alien invasion.<ref name=AC976/>

''In [[New 52]] continuity, their son was born in an alternate Batcave, delivered by [[Thomas Wayne (Flashpoint Timeline)|Batman,]] during [[Convergence (event)| Convergence.]]<ref>{{c|Convergence: Superman Vol 1 2}}</ref>  This was later retconned by the Superman Reborn crossover.''

After Jon's birth, the Kent family moved to a farm upstate, and instead commuted to work in Metropolis.  Jon Kent was raised with no knowledge of his father's dual identity as Superman and was largely kept isolated from many of Superman's associates.

====The Adventures Continue====
At some point, Superman came into conflict with the [[The Elite (Prime Earth)|Elite]], a violent and proactive super-team lead by [[Manchester Black (Prime Earth)|Manchester Black]]. He developed an instinctive distaste for Manchester as a result; Manchester felt similarly.<ref>{{C|Superman and the Authority Vol 1 1}}</ref><ref>{{C|Superman and the Authority Vol 1 4}}</ref>

Over the course of his long career, Superman fought at the side of "one or two" of the historical figures known as [[Arthur Pendragon (Prime Earth)|King Arthur]]. In the process, he rescued the famed 'round table' during the Second Fall of [[Camelot]] and was entrusted with its safekeeping by Arthur himself.<ref>{{C|Superman and the Authority Vol 1 1}}</ref>

Eventually, Clark introduced his clone Conner Kent to his parents, and Connor lived with them and attended Smallville High School for a time before he was lost in Gemworld.<ref name=YJ-V3-3>{{c|Young Justice Vol 3 3}}</ref>

====Imprisoning Luthor====
Superman, Batman, Martian Manhunter, and Wonder Woman would later fight against a team-up of Lex Luthor and [[John Corben (Prime Earth)|Metallo]]. Diana defeated Metallo, unbeknownst to Lex, allowing Martian Manhunter to assume his shape.  Superman then pretended to be affected by 'Metallo's' [[Kryptonite]]. Seeing victory ahead, Lex took a moment to gloat; unbeknownst to him, Manhunter's faux-Kryptonite was a recording device, and his admission of attempted murder was enough evidence to send him to prison.<ref>{{C|Superman: Son of Kal-El 2021 Annual Vol 1 1}}</ref>

Superman later visited Luthor in prison, seeking to discuss the fact that the welder who would link Lex to Metallo's armor plating had gone missing. Instead, Lex offered to place Chess with Superman, who humored him for a short while- until he was informed that the welder had been found, safe but a little worse for wear.  He then left, telling Lex that he wasn't 'playing his game'.<ref>{{C|Superman: Son of Kal-El 2021 Annual Vol 1 1}}</ref>

===[[New 52]]===
{{Image|Superman Prime Earth 0004.jpg|The Man of Tomorrow|left|150px}}
After Barry Allen returned from the [[Flashpoint]], Clark adopted a new costume; a suit comprised of Kryptonian nanotech that would construct itself over his clothes when he needed it.

''In the original New 52 continuity, Clark found this suit aboard Brainiac's ship during his origin and wore it from then on; however, as the result of successive continuity changes, Superman does not begin wearing this suit until much, much later.''

====Death of "Clark Kent" and Fighting Vyndktvx====
{{main|Action Comics: At the End of Days}}
After faking his death as Clark Kent to shake off someone who was investigating too closely into his secret identity, Clark took the alias of Johnny Clark and joined the fire department. Still, he began to regret killing Clark Kent off and sought advice from [[Batman (Bruce Wayne)|Batman]], who promised to deal with it.

After discovering an alien named Captain Comet who foretold of Earth's extinction by alien attacks, Superman managed to fight him off but, in the process, caused his Johnny Clark identity to be thought dead ''as well''. Luckily, Superman's former landlady, Mrs. Nyxly, who knew of his former secret identity as Clark Kent, revealed to him that she was from the [[Fifth Dimension]] and used one of her three wishes she had saved to wipe Clark Kent's death from everyone's minds. Nyxly continued to tell Superman that everything bad that ever happened in his life was the doing of someone ''else'' from her home.<ref name ="AC12">{{C|Action Comics Vol 2 12}}</ref> She explained that [[Vyndktvx (Prime Earth)|Lord Vyndktvx]] from her home dimension had attacked Clark at ''multiple'' points in his history, to the point that certain memories he had might have happened otherwise ''before'' those attacks.<ref name ="AC15">{{C|Action Comics Vol 2 15}}</ref>

Lord Vyndktvx finally made a play for Superman's life in the present with his assembled Anti-Superman Army.<ref name="AC15"/> Finally understanding Vyndktvx's weakness, Superman used a telepathic connection set up by Lois Lane's psychic niece [[Susan Tompkins (Prime Earth)|Susie Thompkins]] to convince everyone in the world to say their own names backwards. The sound of this five-dimensional utterance caused Vyndktvx to retreat to his own dimension, where he was immediately arrested.<ref>{{C|Action Comics Vol 2 18}}</ref>

====H'El on Earth====
{{Main|Superman: H'El on Earth}}
{{Image|Superman_Vol_3_14_Textless.jpg|Superman vs. H'el|left|150px}}
Soon after, the [[Daily Planet]] was bought out by businessman [[Morgan Edge (Prime Earth)|Morgan Edge]], initiating a major change in the reporting style of the Planet. Unhappy with the changes, Clark quit working for the paper.

In a [[The New 52|previous iteration of history]], Clark then discovered that found out that his longtime crush [[Lois Lane (Prime Earth)|Lois Lane]] was now in a serious relationship with another man named [[Jonathan Carroll (Prime Earth)|John Carroll]].<ref>{{C|Superman Vol 3 1}}</ref> However, this event was erased from history when [[Superman: Reborn|Superman's history was revised]].

Later, in his hero identity, Superman was confronted by another Kryptonian survivor called [[H'el (Prime Earth)|H'El]], who wanted to recruit him by force to join his unholy crusade of saving Krypton. After learning that H'el wanted to use the Fortress of Solitude to go back in time to save Krypton by extinguishing Earth's sun, Superman called in the Justice League and his teenage clone [[Kon-El II (Prime Earth)|Superboy]] to fight H'El and prevent the destruction of the solar system.<ref>{{c|Superman Vol 3 15}} - {{c|Superman Vol 3 17|17}}</ref>

====[[Superman Unchained|Unchained]]====
While investigating the shady government group known as [[Ascension (Prime Earth)|Ascension]], Superman uncovered the beast known as [[Wraith (Prime Earth)|Wraith]], an alien with very similar powers to him that was captured and trained by the US government to bring down terrorists. After learning about their common purposes, Superman and Wraith agreed to work together to stop Ascension.<ref>{{c|Superman Unchained Vol 1 3}}</ref>

Superman and Wraith defeated a small faction of Ascension, but Superman was soon betrayed by Wraith, who sought what was known as an Earthstone. Following a colossal battle, Superman defeated Wraith and not long after this, Lex Luthor informed him that Wraith's species was planning to attack Earth and the only weapon that could be used against them would kill Superman.<ref name=SMU8>{{c|Superman Unchained Vol 1 8}}</ref> Superman chose not to use the weapon and confronted the aliens without aid, but Wraith soon redeemed himself by using the weapon against his own species, killing himself in the process.<ref>{{c|Superman Unchained Vol 1 9}}</ref>

<div class="mw-collapsible mw-collapsed" style="width:100%">
====The Psi-War====
<div class="mw-collapsible-content">
{{main|Superman: Psi-War}}
After the battle with H'El, Superman was called to a U.S. Senate hearing to discuss the existence of the Fortress of Solitude. The Senate representatives wanted to investigate the Fortress in order to secure or remove anything that might be considered a threat to the world. However, Superman refused, appealing to his right to privacy and the fact that the Fortress contained technology beyond humanity's control. Clark also became a blogger to make ends meet.<ref>{{c|Superman Vol 3 18}}</ref>

When it was discovered that the villain [[Hector Hammond (Prime Earth)|Hector Hammond]], after falling into a coma, used his telepathy to hide his consciousness within Superman's memories, [[Wonder Woman (Diana Prince)|Wonder Woman]] and [[Orion (Prime Earth)|Orion]] team up to banish Hammond back to his own body and save Superman. For the future, Orion used his Mother Box to grant Superman protection from telepathic assaults.<ref name=SM-3-20/>

Continuing his investigative work, Clark went to Gotham City, where a mysterious Chinese woman gave him, a folder containing info about an urban myth called "the Twenty". However, before he could break the story, Superman was incapacitated by a psychic attack caused by the H.I.V.E. Queen, one of the Twenty.<ref>{{c|Superman Vol 3 22}}</ref>

Superman and Hector Hammond teamed up to stop the Queen, but, upon her defeat and apparent death, a new villain arrived: the [[Roger Hayden (Prime Earth)|Psycho Pirate]], a telepath who knew about Superman's secret identity.<ref>{{c|Superman Vol 3 23}}</ref> The Psycho Pirate lead Superman to the Swarm, a prison for psionic people, explaining that he was also one of the Twenty. The Queen created the Swarm in order to prepare the world for Brainiac's return. With the Queen dead, the Psycho Pirate wanted to release the people trapped in the Swarm, so he trapped Superman in an illusionary world in order to absorb his energy.<ref>{{c|Action Comics Vol 2 24}}</ref>

Lois, having developed psychic powers of her own, freed Superman from the illusions. Knowing they were outmatched against the Psycho Pirate, Superman and Lois joined forces with Hammond and the H.I.V.E. Queen, who had barely survived the Pirate's attack, and defeated the Psycho-Pirate and destroyed his [[Medusa Mask]].

During the battle, Lois learned Superman's secret identity but conveniently fell into a coma from her new power usage. Superman brought Lois to the hospital to be treated and, when she reawakened, she had forgotten the information and lost her powers.<ref>{{c|Superman Vol 3 24}}</ref>
</div>
''Following various revisions to history, it is likely this event was erased from continuity (or at least heavily modified), as it no longer aligns with the histories of many of the characters involved, including that of Lois Lane and the Psycho-Pirate.''

====Krypton Returns====
{{Main|Krypton Returns}}
{{Image|Superman_Vol_3_25_Textless.jpg|Superman and Jor-El|right}}
Superman was dragged into space by a mysterious force called the [[Oracle (Prime Earth)|Oracle]] that told him that H'El had survived and had managed to bring Krypton back to life by altering the timeline, which threatened the entire universe if they did not go back in time to stop him. With the help of [[Faora Hu-Ul (Prime Earth)|Faora]], Superman traveled to Kryptonopolis a few months before Krypton's destruction, where he met his mother Lara.<ref>{{c|Action Comics Annual Vol 2 2}}</ref><ref>{{c|Superboy Vol 6 25}}</ref> Superman witnessed Lara tell Jor-El that she was pregnant,<ref>{{c|Supergirl Vol 6 25}}</ref> but was soon confronted by a version of his father Jor-El from the dystopian future H'El had created, who had also gone back in time to fix the present.

Superman and Jor-El traveled to Krypton's core and fought H'El, and Superman used H'El's own chronal powers against him by freezing him with his super-breath, locking H'El in a state of eternal limbo. Superman and Supergirl were reunited in the present, but they learned that Superboy sacrificed his life to stop H'El. Mourning the loss of their friend, Superman and Supergirl returned to Earth.<ref>{{c|Superman Vol 3 25}}</ref>

====Trinity War and Forever Evil====
{{Main|Trinity War|Forever Evil}}
In the hopes of ridding the world of evil, [[Pandora (Prime Earth)|Pandora]] tried to have members of the [[Justice League (Prime Earth)|Justice League]] open [[Pandora's Box|her box]], believing that their virtuousness would prevent the calamity that occurred when ''she'' opened it.<ref>{{C|Trinity of Sin: Pandora Vol 1 1}}</ref> However, this backfired, and the active Justice Leagues took sides, fighting over the "correct" method to open the box.<ref>[[Trinity War]]</ref>

While trying to understand the escalating tension of opening the box, Superman and Wonder Woman were ambushed by [[Arthur Light (Prime Earth)|Dr. Light]] and, during their fight, Superman accidentally killed Light with his heat vision.<ref>{{c|Justice League Vol 2 22}}</ref> Superman asked to be locked away for his outburst but started getting sick in captivity.<ref>{{c|Justice League of America Vol 3 6}}</ref> Upon investigating further, Firestorm, the Atom and Element Woman discovered that there was Kryptonite in Superman's nervous system, which caused him to attack Dr. Light unintentionally.<ref>{{c|Justice League Vol 2 23}}</ref> However, before they could save him, the [[Crime Syndicate (Earth 3)|Crime Syndicate]] came to Earth from their world and trapped the Justice League inside the [[Firestorm Matrix]].

After being freed by Lex Luthor and his [[Injustice League (Prime Earth)|group of rebellious criminals]], Superman was eventually saved by Luthor, who was volunteered to remove the Kryptonite from his brain.<ref>{{c|Forever Evil Vol 1 7}}</ref> For his heroism during the crisis, Lex Luthor was accepted as a member of the [[Justice League (Prime Earth)|Justice League]].

====Doomed====
{{Main|Superman: Doomed}}
{{Image|Action_Comics_Annual_Vol_2_3_Textless.jpg|SuperDoom|right|150px}}
After discovering [[Doomsday (Prime Earth)|Doomsday]] wreaking havoc and harboring a dangerous toxin within his body, Superman ripped the beast in half and used his super-breath to inhale the toxin to stop it from spreading.<ref>{{c|Superman: Doomed Vol 1 1}}</ref> After recovering, Superman realized that he was slowly turning into a Doomsday-like creature himself.<ref>{{c|Action Comics Vol 2 31}}</ref> Wonder Woman tried to help him, but the transformation became stronger<ref>{{c|Superman/Wonder Woman Vol 1 8}}</ref> and Superman had no choice but to turn himself to the authorities.<ref>{{c|Batman/Superman Vol 1 11}}</ref> Superman was locked in a special facility but he soon was needed by the Teen Titans to defeat an enemy, which he eliminated ruthlessly in his increasingly monstrous state.<ref>{{c|Superman Vol 3 31}}</ref>

The military declared Superman an enemy of the state and they sent Metallo and Atomic Skull to eliminate him. Steel tried to help Superman fight the enemies, but, after a suicide attack with kryptonite by Metallo, Superman became weak and his transformation into "SuperDoom" was accelerated.<ref>{{c|Action Comics Vol 2 32}}</ref> Wonder Woman then took Clark to outer space where he could escape into its depths until they found a way to save him.<ref>{{c|Superman/Wonder Woman Vol 1 9}}</ref>

While in space, SuperDoom tried to help an alien colony but lost control and attacked their leader.<ref>{{c|Action Comics Vol 2 33}}</ref> He was then warned about an incoming invasion of Earth by Brainiac's forces and he confronted its commander, a [[Zor-El (Prime Earth)|cyborg with Superman's appearance]].<ref>{{c|Superman/Wonder Woman Vol 1 10}}</ref> After a brief fight between them, SuperDoom returned to Earth and joined the Justice League to fight Brainiac's forces and after their victory, the Justice League removed the kryptonite from Earth's atmosphere, making Superman temporarily change back to normal.<ref>{{c|Superman/Wonder Woman Annual Vol 1 1}}</ref><ref>{{c|Action Comics Annual Vol 2 3}}</ref>

When Brainiac's forces returned with reinforcements, Superman attempted to turn back into SuperDoom to fight them. Unfortunately, Superman lost control of his body completely to the virus.<ref>{{c|Superman/Wonder Woman Vol 1 11}}</ref> Defeating Brainiac, Lois Lane helped Superman steal Brainiac's powers and finally cure Superman of the Doomsday virus. Once free, the hero captured Brainiac and took him inside a black hole,<ref>{{c|Superman: Doomed Vol 1 2}}</ref> returning back to Earth sixty days later.<ref>{{c|Action Comics Vol 2 35}}</ref>

====Men of Tomorrow====
{{Image|Superman Vol 3 35 Textless Romita, Jr. Variant.jpg|Superman and Ulysses|left|150px}}
During an alien attack on Metropolis, Superman was assisted by a stranger called [[Neil Quinn (Prime Earth)|Ulysses]] and together they defeated [[Klerik (Prime Earth)|Klerik]], one of the alien soldiers. Superman then learned that Ulysses believed himself to be the last survivor of Earth.<ref>{{c|Superman Vol 3 32}}</ref> With Superman's help, Ulysses found the truth about his past,<ref>{{c|Superman Vol 3 33}}</ref> but they were soon confronted by "The Machinist", a man responsible for creating destructive robots.<ref>{{c|Superman Vol 3 34}}</ref> The criminal managed to escape and shortly afterwards, Ulysses revealed his plan of taking millions of people to his home planet, "The Great World" in order to save mankind from what he believed was Earth's doom.<ref>{{c|Superman Vol 3 35}}</ref> This lead to a confrontation between Superman and Ulysses, with the latter winning and taking Superman prisoner.<ref>{{c|Superman Vol 3 36}}</ref>

On their way to Ulysses' home planet, there was another confrontation between Superman and Ulysses<ref>{{c|Superman Vol 3 37}}</ref> and the fight was brought to Metropolis, where Superman unleashed a new power, he didn't know he had: a super flare of solar energy, beating Ulysses for good. Unfortunately, the solar flare drained his powers for a day, leaving him as a normal human until they slowly began to come back.<ref name=SM-3-38>{{c|Superman Vol 3 38}}</ref>

====Blackmailed by HORDR====
At the ''Daily Planet'', Clark started to get text messages from an unnamed source that gave him tips on criminal operations that he was able to stop as Superman. Unfortunately, he got more messages from the unnamed source, who revealed they knew Clark's secret identity and forced him to capture a woman called Condesa. Realizing that Condesa knew about the person blackmailing him, Clark chose to team up with her instead.<ref name=SM-3-41>{{c|Superman Vol 3 41}}</ref>

Condesa told Clark about the organization that had been stalking her called HORDR, whose leader HORDR_ROOT was behind Superman's blackmailing. As Superman went after HORDR, he was forced to reveal his secret identity to Lois Lane and, while using his new solar flare power to destroy the HORDR base, was infected with a substance by their leader that neutralized his powers.<ref>{{c|Superman Vol 3 42}}</ref>

Much later, as Clark recovered, HORDR_ROOT revealed he survived their encounter and further blackmailed Superman into working with him. Clark begrudgingly agreed to play along and went to a secret facility where HORDR forced him to use his solar flare so HORDR could analyze it. Lois, not wanting Superman to suffer through the blackmail any longer, leaked his secret identity as Clark Kent to the world, as well all the information she had on HORDR. Unfortunately, though HORDR_ROOT no longer had any leverage, Superman, still weakened, was turned into public enemy #1.<ref name=SM-3-43>{{c|Superman Vol 3 43}}</ref>

====Truth and Savage Dawn====
{{Image|Action_Comics_Vol_2_41_Textless.jpg|Public, Depowered, and Dying|150px|left}}
Because of his outed identity, Clark was fired from the Daily Planet by Perry White, who felt betrayed that Clark kept such an important secret from him.<ref>{{c|Superman Vol 3 44}}</ref> And, because he no longer had the powers he usually had, Superman was unable to return to the Fortress of Solitude and his advanced Kryptonian costume didn't recognize him anymore. Clark was forced to change his image - he cut his hair, improvised his signature symbol with a Superman merchandise t-shirt, and tore up his cape to fashion himself knuckle bandages. Superman still attempted to help the people of Metropolis even in his weakened state.

The mastermind behind Superman's power loss was revealed to actually be [[Vandal Savage (Prime Earth)|Vandal Savage]], who hijacked the Stormwatch space carrier and captured the Justice League.<ref name=SM-3-43/> To stop Savage, Steel provided Superman with a Kryptonite-powered suit of armor. Superman and the [[Justice League United (Prime Earth)|Justice League United]] stormed the Watchtower, but Vandal tore the Fortress of Solitude off the ground in his plan to gain power from the comet that made him immortal in the first place.<ref>{{c|Superman/Wonder Woman Vol 1 26}}</ref> Vandal used the comet's power to give superpowers to the thousands of illegitimate children he had fathered across the millennia, turning them into an army he could throw at Superman and his allies.<ref name=SM-3-49>{{c|Superman Vol 3 49}}</ref>

Superman lead Earth's remaining superheroes into attacking the Stormwatch carrier and freed the Justice League. Superman attempted to stop Savage, but the villain threw him into the Fortress of Solitude, which finally acknowledges him as Kal-El of Krypton, healed his body and restored his uniform.<ref>{{c|Action Comics Vol 2 50}}</ref>

Fully powered again, Superman attempted to stop the fight, as Vandal manipulated his children into exposing themselves to the comet's dangerous power. Unfortunately, Superman's pleas fell on deaf ears, and their bodies collapsed from the power they had absorbed.<ref>{{c|Superman/Wonder Woman Vol 1 27}}</ref> Vandal offered for Superman to join his side and rule Earth, but Superman refused and broke the comet off Earth's orbit, throwing it and Vandal into deep space.<ref>{{c|Superman Vol 3 50}}</ref>

====Darkseid War====
{{main|Justice League: The Darkseid War}}
{{Image|Superman_Prime_Earth_0037.jpg|Superman as the God of Strength|right|150px}}
When the daughter of Darkseid named [[Grail (Prime Earth)|Grail]] signaled an impending war between [[Darkseid (Prime Earth)|her father]] and the [[Mobius (Sixth Dimension)|Anti-Monitor]], Superman and Lex Luthor, still a member of the Justice League, attempted to avert the disaster. However, the two men were betrayed by Lex's sister Lena, who used a Mother Box to teleport them to Apokolips.<ref>{{c|Justice League Vol 2 41}}</ref> Once there, Superman and Lex were forced to work together in order to survive, especially because Apokolips was far away from a yellow sun, meaning that Superman's powers were depleting quickly.<ref>{{c|Justice League Vol 2 42}}</ref> As they were surrounded by Parademons, Lex, in a desperate attempt to save the Man of Steel, threw Superman into one of Apokolips' fire pits, thinking its energy would revitalize him.

Unfortunately, Apokolips' energy darkened Superman's body and mind, turning him into the God of Strength.<ref>{{c|Justice League Vol 2 43}}</ref> With his new power, the God of Strength destroyed the enemies surrounding him and Lex's armor<ref>{{c|Justice League Vol 2 44}}</ref>, leaving Luthor alone on Apokolips while he himself returned home.<ref>{{c|Justice League Vol 2 45}}</ref><ref>{{c|Justice League: The Darkseid War: Superman Vol 1 1}}</ref>

When Superman reunited with the Justice League in his altered state, Wonder Woman was able to use her [[Lasso of Truth]] to bring Superman to his senses.<ref>{{c|Justice League Vol 2 47}}</ref> As the war between Darkseid and the Anti-Monitor continued to wreak havoc on Earth,<ref>{{c|Justice League Vol 2 48}}</ref> Superman joined the fight and briefly staggered the Anti-Monitor by blasting him with all the dark energy in his body, turning himself back to normal. The war was halted when Mobius was unexpectedly killed by Steve Trevor, who had been infected with the Anti-Life Equation by Grail.<ref name=JL-2-49>{{c|Justice League Vol 2 49}}</ref>

In the final fight, Lex and Steve were stripped of their powers to resurrect the fallen Darkseid as a baby. With all the enemy combatants either dead or scattered, the war was over. Unfortunately, despite expelling the Apokoliptian dark energy during the fight, the damage done to Superman's cellular structure was irreversible. According to Batman, Superman was dying.<ref>{{c|Justice League Vol 2 50}}</ref>

====Final Days of Superman====
{{main|Superman: The Final Days of Superman}}
As a result of cosmic upheaval of [[Flashpoint]] and [[Rebirth]], Earth-0 had inadvertently become home to two versions of both Clark Kent and Lois Lane, split from one another during the creation of the post-Flashpoint timeline.  One group appeared to be 'native' to the post-Flashpoint timeline; the other pair seemed to be remnants of the timeline's [[New Earth|original history]] prior to its alteration.  Though they existed apart for a time, eventually, their fates converged.

After being depowered by Vandal Savage, the events of the Darkseid War, and his fight with the Kryptonian god [[Rao (Prime Earth)|Rao]], Superman learned that his death was imminent and asked for Batman's help to find him a successor.<ref>{{c|Batman/Superman Vol 1 31}}</ref> As part of his legacy, Superman told the whole truth to Lois Lane and asked her to publish everything in the Daily Planet in order to have his history and identity revealed to the world before his death.<ref>{{c|Superman Vol 3 51}}</ref> Superman told the truth to Diana, and they shared a moment before they were needed back in Metropolis, where the [[Denny Swan (Prime Earth)|"Energy Superman"]] they had caught days earlier had escaped Stryker Island and was destroying everything.<ref>{{c|Superman/Wonder Woman Vol 1 29}}</ref>
{{Image|Death_of_Prime_Earth_Superman_001.PNG|2nd Death of Superman|150px|right}}

Superman, Batman and Wonder Woman together tried to stop the Energy Superman and, while doing so, were assisted by the mysterious "[[Kal-El (New Earth)|Superman in black]]", who chose to remain unidentified for the time being. Superman took the Energy Superman to space, where he used his solar flare power to finish the threat, and the enemy evaporated in the cosmos. Superman's body fell down to Earth and, as he was surrounded by his friends, he asked the "Black Superman" to take his mantle and become the ''New'' Superman of ''Earth''. After this, the other Superman died and turned to ashes.<ref>{{c|Superman Vol 3 52}}</ref>

This Superman's powers would be transferred to both [[Lana Lang (Prime Earth)|Lana Lang]] and [[Lois Lane (Prime Earth)|Lois Lane]], who were positioned closest to him when his energy gave out. They would both take the alias [[Superwoman]], but Lana would later lose the powers and that Lois would die in action.</div>

''Within the history of [[Dawn of DC]], the Lois Lane who acted as Superwoman is described in computer database of [[John Henry Irons (Prime Earth)|John Henry Irons]] as an alternate version of Lois Lane (a reversal of the original nature of the storyline) and as a 'remnant' of energies resulting from the [[Rebirth]] event.  It is likely the same applies to Superman.''<ref name=SW1>{{c|Steelworks Vol 1 2}}</ref>

===[[DC Rebirth]]===
====Superman Reborn====
{{main|DC Rebirth}}
Not long after, [[Mister Mxyzptlk (New Earth)|Mr. Mxyzptlk]] appeared, having been imprisoned by [[Jor-El (Prime Earth)|Mr. Oz]].  He initially appeared as Clark Kent, much to the confusion of all, as it was common knowledge that Superman and Clark Kent were one in the same.  He began to harass Superman and revealed that both the New Earth and Prime Earth Supermen were actually two halves of the true, complete Superman, who was split during the [[Flashpoint]] incident. The same was true for Lois.<ref>{{c|Superman Vol 4 19}}</ref> These alternates versions of Clark and Lois merged with the deceased essences of their Prime Earth counterparts, and each became single, complete Prime Earth versions - fixing their conflicting histories and bringing the world back to normal.<ref name=AC976>{{c|Action Comics Vol 1 976}}</ref>

Afterwards, memories of this incident were vague. However, Mxyzptlk's impersonation of Clark Kent did have a beneficial side effect; because he told others that the revelation of Superman's secret identity was a farce, a ruse needed to draw out an enemy, and because he was photographed alongside Superman, the public and many others were convinced that Superman was in fact ''not'' Clark Kent, restoring Superman's secret identity.
 
====Revenge====
{{main|Superman: Revenge}}
As Clark and Lois got ready to move back to Metropolis from their rural home in the country, the new [[Henry Henshaw (Prime Earth)|Hank Henshaw]] put together a new [[Superman Revenge Squad]].<ref>{{c|Action Comics Vol 1 979}}</ref> Stealing a mystical item from Superman's Himalayan backup Fortress, Henshaw turned himself back into Cyborg Superman and allied himself with Superman's greatest foes to finally destroy him.<ref>{{c|Action Comics Vol 1 982}}</ref>

Superman was forced to retreat, taking his family to the Fortress of Solitude. The group of villains assaulted the Fortress before he had time to come up with a plan, but the [[Superman Family]] - [[Kara Zor-El (Prime Earth)|Supergirl]], [[Lana Lang (Prime Earth)|Superwoman]], [[John Henry Irons (Prime Earth)|Steel]], [[Kong Kenan (Prime Earth)|Super-Man]] and [[Alexander Luthor (Prime Earth)|Lex Luthor]] (also [[Superman]]) arrived, ready to help him.<ref name=AC983>{{c|Action Comics Vol 1 983}}</ref>

During the ensuing brawl, Zod managed to send most of Superman's allies to the [[Phantom Zone]] and freed his wife Ursa and their son Lor-Zod. Superman and his family successfully defeated the rest of the Squad and freed Superman's friends. As punishment, Superman left the defeated Cyborg Superman in the Phantom Zone.<ref>{{c|Action Comics Vol 1 984}}</ref>

====The Oz Effect====
{{Image|Action_Comics_Vol_1_989_Textless.jpg|The truth of Mr. Oz|150px|right}}
After addressing several emergencies all over the world, Superman found the perpetrator of these disasters to be the mysterious force that had been watching him for years - Mr. Oz, who teleported Superman to the Fortress of Solitude and revealed himself to be Superman's Kryptonian father [[Jor-El (Prime Earth)|Jor-El]].<ref>{{c|Action Comics Vol 1 987}}</ref> Jor-El told his son about how he had survived - a [[Jonathan Osterman (Watchmen)|mysterious entity]] rescued him from Krypton moments before it exploded and brought him to Earth to witness humanity's darkest impulses. Horrified by these atrocities, Jor-El became Mr. Oz and vowed to correct the decision he had made when Kal was a baby and escort him to a better planet to live.<ref>{{c|Action Comics Vol 1 988}}</ref> Superman refused to abandon humanity though and stood by his mission to help whenever he was needed.<ref>{{c|Action Comics Vol 1 989}}</ref>

Oz set his sights on Superman's family and made contact with Lois and Jon, hoping that, by convincing them to rebuke humanity, Kal would too.<ref name=AC990>{{c|Action Comics Vol 1 990}}</ref> Superman, furious at his father's nagging, physically fought him and shattered the man's staff, which had been mind-controlling him to do his evil actions. Back to his senses, Jor-El begged his son to do the right thing before being teleported away by the same entity that saved him before.<ref>{{c|Action Comics Vol 1 991}}</ref>

====Back to Krypton's Past====
Saddened by these recent events, Superman investigated if Mr. Oz was really his father by using the Flash's [[Cosmic Treadmill]] to travel back in time to Krypton's past.<ref name=AC992>{{c|Action Comics Vol 1 992}}</ref> Superman reached Krypton, where he stumbled upon time-traveler [[Michael Carter (Prime Earth)|Booster Gold]] and [[Skeets (Prime Earth)|Skeets]], who informed the Man of Steel that his presence had distorted the timeline so that the Kryptonians began properly making preparations to evacuate their dying planet.<ref>{{c|Action Comics Vol 1 993}}</ref> Booster attempted to forcibly return Superman to his original timeline, but one of Krypton's Eradicators damaged his time sphere, and they became trapped in [[25th Century]] Gotham City, Booster's birthplace.<ref>{{c|Action Comics Vol 1 994}}</ref>

{{Image|Action_Comics_Vol_1_994_Textless.jpg|Altering Krypton's history with time travel|150px|left}}
Gaining access to a [[Time Sphere]], Booster theorized they could now visit Krypton without risking further damage to the timeline. However, the Eradicator then, coincidentally, also damaged the time sphere, and they crash-landed and were captured on a mysterious planet that worshiped General Zod.<ref>{{c|Action Comics Vol 1 995}} - {{c|Action Comics Vol 1 996|996}}</ref> The heroes were sentenced to be executed but, before Zod and his family got the chance, Superman, Booster and Skeets caused an uprising of slaves on the planet that allowed them to escape.

Skeets reconstructed the time sphere, giving them the chance to return to the present. Before they did though, Superman convinced Booster to let him see the final moments of Krypton and, upon their arrival, watched as a mysterious blue light took Jor-El away, confirming that he was indeed Mr. Oz.

The heroes returned to the present where, in Superman's absence, his family would have died, but Booster Gold secretly rewound time to give Superman enough time to save them.<ref>{{c|Action Comics Vol 1 998}}</ref>

====Invasion of the [[Dark Multiverse]]====
{{main|Dark Nights: Metal}}
Through unknown circumstances, the Justice League (Superman included) was captured by [[Mongul (Prime Earth)|Mongul]], who had forced the League into gladiatorial combat against machines built by a captive [[Hiro Okamura (Prime Earth)|Toyman]]. Fortunately, Batman had figured out a way to help the League escape from their predicament, defeat Mongul and return to Earth. As soon as they returned, however, they discovered that a large mountain had suddenly grown in the middle of Gotham. Their investigations on the mountain led them to [[Kendra Saunders (Prime Earth)|Kendra Saunders]], who had also been investigating the mountain, believing it to be related to a mysterious region of space known as the [[Dark Multiverse]], a region of space far removed from the limits of Universe-0. Kendra told the League that a mysterious evil from the Dark Multiverse would use Batman to unleash a great crisis upon their world and attempted to apprehend Batman, who created a distraction to escape.<ref name="DKM-1">{{c|Dark Nights: Metal Vol 1 1}}</ref>

{{Image|Dark Nights Metal Vol 1 2 Textless.jpg|150px|right|The Trinity faces the horrors of the Dark Multiverse.}}
As the League searched for Batman all over the world, Superman refused to believe that his friend would be the catalyst for the Dark Multiverse's victory and instead believed that Batman could be reasoned with. Superman and the League battled through several distractions left in place by Batman until he and Wonder Woman cornered Batman in the Temple of [[Hath-Set (Prime Earth)|Hath-Set]], the place prophesized to be the dark army's entrance to Universe-0. Unfortunately, the [[Court of Owls (Prime Earth)|Court of Owls]] manipulated Batman into opening the gate to the Dark Multiverse, allowing [[Barbatos (Dark Multiverse)|Barbatos]], king of the Dark Multiverse, and his seven [[Dark Knights (Dark Multiverse)|Dark Knights]] entry into Universe-0. Barbatos and his Dark Knights threw Batman into the Dark Multiverse and began draining the life-forces of Superman and Batman, intending to use them to begin his conquest of the regular Multiverse.<ref>{{c|Dark Nights: Metal Vol 1 2}}</ref>

For an entire week, Barbatos held the captured heroes into dark towers hanging above the ruins of Metropolis while subjecting Superman and Wonder Woman into terrible nightmares that convinced the heroes that they had triumphed over the dark forces of Barbatos, only to reveal that Barbatos had let them win. Against Wonder Woman's warnings, Superman attempted to defeat Barbatos all by himself, only to be ambushed by one of his Dark Knights, the [[Bruce Wayne (Earth -1)|Devastator]]. Superman was only saved by the timely interference of the Flash and taken to the Oblivion Bar, one of the last safe places on Earth. As the heroes discussed their plans to stop the invasion, Clark revealed that Bruce had sent him a warning through Barbatos' nightmares and believed he could journey into the Dark Multiverse and save him. With help from Flash and Steel, Superman traveled to the Fortress of Solitude and used modified his [[Phantom Zone Projector]] to create a portal to the Dark Multiverse. Although Superman managed to reach Batman, he was captured by the evil Supermen of the Dark Multiverse.<ref name="DKM-3">{{c|Dark Nights: Metal Vol 1 3}}</ref>

Using the last of his strength, Batman saved Clark from the dark Supermen and asked [[Daniel Hall (Prime Earth)|Dream]] of the [[Endless]] for help in stopping Barbatos. The Prince of Stories told the World's Finest about the origins of Barbatos and the Dark Multiverse and revealed that there was a way to defeat the forces of darkness. The heroes needed to travel to the World of Forge, at the heart of creation, and use the metal from the forge to create weapons that could defeat Barbatos. Upon reaching the forge, however, the heroes found that the forge had gone dark and were ambushed by [[Carter Hall (Prime Earth)|Carter Hall]], who had been transformed into Barbatos' dark dragon.<ref name="DKM-4">{{c|Dark Nights: Metal Vol 1 4}}</ref>

As Batman attempted to reason with Carter, Superman listened to Barbatos wail into the skies, intending to accelerate their universe's fall into the Dark Multiverse. The heroes distracted Carter long enough to find a spark of hope within the Forge of Worlds, just in time for the dark metals of the Dark Multiverse to start flooding the Forge.<ref>{{c|Dark Nights: Metal Vol 1 5}}</ref> Fortunately, Wonder Woman reached the lowest depths of the Forge and found Superman and Batman with the last remnants of Tenth Metal, which they used to forge armors for the entire Justice League. As the League fought Barbatos and his Dark Knights, Superman personally defeated the Devastator while using the mental powers of the Tenth Metal to bind Barbatos in cosmic chains and return the Multiverse to normal. The outcome of the battle, however, shattered the Source Wall, leaving the positive Multiverse exposed to the mysteries and dangers of whatever lied beyond.<ref>{{c|Dark Nights: Metal Vol 1 6}}</ref>

{{Image|Justice League No Justice Vol 1 1 Textless Wraparound.jpg|left|150px|Superman as a member of Team Mystery.}}
Peace would be short-lived, however, as Brainiac attacked Earth, captured as many superhumans as he could (Superman included) and forcibly brought them to his homeworld, Colu, which was under attack by cosmic entities known as the [[Omega Titans]]. He explained to his prisoners that he needs their help to save Colu from the Titans, who were awakened by the breaking of the Source Wall. To stop them, the superhumans of Earth would need to find the four energies connected to each of the Titans (Mystery, Wonder, Entropy and Wisdom) and achieve a balance betweent them, which would give the Titans no reason to consume Colu. Superman was forced to agree to Brainiac's condition and was placed with a team of fellow superhumans that embodied Mystery the same way he did. As Brainiac attempted to explain his plan, however, he suffered a seizure and died in Superman's arms.<ref>{{c|Justice League: No Justice Vol 1 1}}</ref>

As the Justice League made planetfall, they found 4 mysterious trees made of pure information that allowed the Titans to discover which of the 4 energies was dominant on Colu.  Team Mystery made its way to the Tree of Mystery, a place that held countless miniaturized cities like Kandor. <ref>{{c|Justice League: No Justice Vol 1 2}}</ref> Martian Manhunter and Starro succeeded in awakening the prisoners from their cities and the other Justice Leagues attempted to balance the energies of the trees. Unfortunately, the Omega Titans devastated Colu and Starro sacrificed his life and bought the Leagues enough time to escape.<ref>{{c|Justice League: No Justice Vol 1 3}}</ref>

The Justice League returned to Earth, only to find the Omega Titans already carving up a path of destruction throughout the planet thanks to a Cosmic Seed Brainiac had planted on Earth to lure the Titans away from Colu. To save the Earth, the League charged the Seed with the 4 energies of Earth and fired it at the Entropy Titan, infusing it with the 4 energies. The other Titans, attracted to the Seed, ate their own brother, and departed from Earth. Although Earth was saved, the Justice League knew that they needed to prepare against whatever challenges may come next. The rules of the universe had been broken and the universe was changing.<ref>{{c|Justice League: No Justice Vol 1 4}}</ref>

====Rampage of Rogol Zaar====
Shortly after the city of Metropolis paid him a tribute for years of heroism,<ref name=AC_1000>{{c|Action Comics Vol 1 1000}}</ref> the Kent family received an unexpected visit from [[Jor-El (Prime Earth)|Mr. Oz]], who asked the Kents to let him take their son Jon on a journey across the galaxy so he could learn about his Kryptonian heritage. Clark initially refused, but Jon insisted he wanted to go. Lois, supporting her son, also agreed to stay at Jon's side for the journey so she could look after him. Clark was reluctant to see his family go, but nevertheless accepted their decision.<ref name="MOS-V2-6">{{c|The Man of Steel Vol 2 6}}</ref>

{{Image|The_Man_of_Steel_Vol_2_6_Textless.jpg|Against the man that killed Krypton|150px|right}}
Afterwards, a mysterious villain broke into the Fortress while Superman wasn't there, leveled the place and, most heinously, smashed the Bottle City of [[Kandor]].<ref>{{c|The Man of Steel Vol 2 3}}</ref> Superman and Supergirl tracked the murderer down and discovered his identity to be an alien named [[Rogol Zaar (Prime Earth)|Rogol Zaar]], who claimed to have been the one to destroy Krypton and wanted to finish his mission of killing all Kryptonians. Superman engaged Rogol all the way to the Moon but was defeated. Rescued by his cousin and healed by the Justice League, Superman figured out, if Rogol intended to cleanse the Kryptonian "plague", his next move would be to blow up the Earth.<ref>{{c|The Man of Steel Vol 2 5}}</ref> Superman and Supergirl worked together to defeat Rogol and trapped him inside the Phantom Zone.

Unfortunately, with the Fortress destroyed, Superman could not communicate with his family and make sure they were safe. Supergirl also decided to leave Earth in order to find out more about Rogol Zaar's origins, and the Justice League held a memorial service for the murdered Kandorians.<ref name="MOS-V2-6"/>

====Earth in the Phantom Zone====
{{Image|Superman_Vol_5_2_Textless.jpg|Trapped in the Phantom Zone.|150px|left}}
After constructing a new Fortress of Solitude in the [[Bermuda Triangle]], Superman had a conversation with [[J'onn J'onzz (Prime Earth)|Martian Manhunter]], who told Superman to consider the possibility of becoming a leader to the global community, so he could lead Earth and humanity to a better tomorrow. However, before he could think on the idea, Clark realized the Earth had somehow been teleported into the Phantom Zone.<ref>{{c|Superman Vol 4 1}}</ref> He later learned the ordeal was an accident caused by scientists at [[S.T.A.R. Labs]], but, unfortunately, the opportunity allowed Rogol Zaar to attack the Earth yet again.<ref>{{c|Superman Vol 4 3}}</ref>

S.T.A.R. Labs initiated a plan to restore the Earth, which involved the [[Raymond Palmer (Prime Earth)|Atom]] miniaturizing the entire planet to make it small enough for the Flash to pass it through the Phantom Zone projector. The plan succeeded and the Earth was returned to its rightful place in the Solar System, but Superman was left behind to fight Rogol alone in the Phantom Zone.<ref>{{c|Superman Vol 4 4}}</ref> With no solar energy to replenish his powers, Superman was badly outmatched, until he was rescued by his enemy General Zod, who agreed to a temporary truce to take revenge on the man who killed Krypton.<ref>{{c|Superman Vol 4 5}}</ref> The two Kryptonians fought the villain together until Superman was abruptly beamed out of the Phantom Zone by his allies, leaving Zod and Rogol Zaar to finish their match alone.<ref name="S-V5-6">{{c|Superman Vol 5 6}}</ref>

====Space Wars and the Birth of the United Planets====
Superman's family returned from their space adventure with [[Zor-El (Prime Earth)|Oz]],<ref>{{c|Action Comics Vol 1 1004}}</ref> but Jon was practically unrecognizable. The science of travelling through space had aged Jon up by seven Earth years, returning as a teenager.<ref name="S-V5-6"/> Superman listened to his family's adventure stories,<ref>{{c|Superman Vol 4 7}}</ref> but, before long - would you believe it <!-- let me have some fun - Harold --> - Rogol Zaar returned ''again'', this time working ''with'' [[Dru-Zod (Prime Earth)|General Zod]]. Superman and his Kryptonian family formed a team to finally take down Rogol Zaar once and for all, with Kal and his father going to the ruins of Krypton to look for clues about the villain.<ref>{{c|Superman Vol 5 12}}</ref>

{{Image|Superman Vol 5 14 Textless.jpg|Kryptonians united|150px|right}}
While on their trip, Jor-El told his son, while on Krypton, he was a member of the Circle, a secret cabal of intergalactic rulers that controlled the galaxy from the shadows. After helping the Circle ensure their dominance across the galaxy, Jor-El expected the Circle to help him save Krypton from its imminent destruction, but they refused. Superman was disappointed by his father's morally gray past.<ref>{{c|Superman Vol 5 13}}</ref>

Rogol Zaar ambushed the heroes, but, fortunately, Zod betrayed Rogol and gave the Family of Steel the opportunity they need to incapacitate Rogol.

When the battle was finished, Superboy, wishing there was better peacekeeping throughout the galaxy, was visited by the [[Legion of Super-Heroes (Post-Rebirth)|Legion of Super-Heroes]] from the [[31st century]], who asked for his membership.<ref>{{c|Superman Vol 5 14}}</ref> Although Superman was nervous at the thought of losing his son yet again, he chose to respect Jon's decision, no matter what it was. Jon deferred the decision for some time, until he decided he would accept the offer and go to the future.

Superman was also informed by [[Adam Strange (Prime Earth)|Adam Strange]] that, due to his past crimes and instigation of the time steam, the nations of the galaxy had ruled that Jor-El was to be sent back in time to die on Krypton like he was supposed to. Kal was devastated, but the sentence was already carried out and there was nothing he could do. Jor-El was returned to Krypton, moments before destruction, where he saw his younger self send Kal on his mission of hope across the cosmos, whispering to himself assuredly, that his son 'did it'. In the aftermath, Zod and his family informed the Kents of their decision to create a "New Krypton" in order to save their race and formally invited the House of El to join them if they ever desired.<ref>{{c|Superman Vol 5 15}}</ref>

{{Image|Superman_Prime_Earth_0042.jpg|Superman speaks the truth|150px|left}}
Once he returned to Earth, Superman talked with Supergirl<ref>{{c|Superman Vol 5 17}}</ref> and Adam Strange and decided to reveal his secret identity to Perry White and Jimmy Olsen, Clark's closest friends.<!-- Because that went sooo well with Perry last time --> Superman then called a press conference and revealed his identity to the entire world.<ref>{{c|Superman Vol 5 18}}</ref> As the people of Earth displayed different reaction to Superman's truth<ref>{{c|Superman: Heroes Vol 1 1}}</ref><ref>{{c|Superman: Villains Vol 1 1}}</ref>, Superman himself worked towards protecting the United Planets from an attack by Mongul, who saw the young interstellar nation as a threat to his own galactic empire.<ref>{{c|Superman Vol 5 19}}</ref> After a difficult struggle in which Mongul tried to devastate the Earth<ref>{{c|Superman Vol 5 21}}</ref>, Superman defeated his opponent and was temporarily elected by the United Nations to be Earth's representative in the galactic stage.<ref>{{c|Superman Vol 5 22}}</ref>

{{Image|Superman Prime Earth 0049.jpg|Superman uses the Helmet of Fate|150px|right}}
After the battle with Mongul, Superman wondered whether or not revealing his secret identity to the people of the world was the right choice, so he consulted [[Khalid Nassour (Prime Earth)|Khalid Nassour]], the latest wizard to hold the mantle of [[Doctor Fate]],<ref>{{c|Superman Vol 5 23}}</ref> to verify whether some magical influence or a demonic possession pushed Clark towards revealing his identity. Superman truly understood that the choice was entirely his own and thus helped Khalid to defeat the Lord of Chaos [[Xanadoth (Prime Earth)|Xanadoth]] in a battle that saw Superman briefly use the Helmet of Fate.<ref>{{c|Superman Vol 5 24}}</ref>

Superman's greatest challenge as ambassador for Earth came in the form of Eisno Alkor, an alien super-soldier from planet Synmar, who was inspired by Superman to end his homeworld's isolationism and forcibly bring it into the galactic community. Realizing that Alkor had conquered his own homeworld in a violent revolution, Superman helped the United Planets and Synmar's resistance movement to depose Alkor and bring order back to Synmar.<ref>{{c|Superman Vol 5 28}}</ref>

====Justice/Doom War====
A mysterious object known as the [[Totality]] landed on the Nevada desert and the Justice League, who had just elected the Martian Manhunter as their new chairman, set out to find it. <ref>{{c|Justice League Vol 4 1}}</ref> Superman and Martian Manhunter's bodies were deemed strong enough to withstand the Totality's energies without negative effects, which is why Batman miniaturized himself to enter Clark's body so he could analyze the Totality. <ref>{{c|Justice League Vol 4 2}}</ref> Unfortunately, while Superman and Martian Manhunter went on to explore the Totality's outer structure, Lex secretly infiltrated Superman's body and used it to harness the Totality's power.<ref>{{c|Justice League Vol 4 4}}</ref> Despite Batman's best efforts, Lex escapes with the Totality, but not before relinquishing control of Superman's body.<ref>{{c|Justice League Vol 4 7}}</ref>

Afterwards, the Justice League dedicated itself towards locating the Legion of Doom's headquarters, but they had to put that search on hold when the [[Triumvirate of Sea Gods]] started their invasion on Earth.<ref>{{c|Justice League Vol 4 10}}</ref> During this event, Metropolis was flooded and Superman was attacked by [[Drogue (Prime Earth)|Drogue]] of the Triumvirate but he was saved by Flash. <ref>{{c|Justice_League/Aquaman:_Drowned_Earth_Vol_1_1}}</ref> Joined by Mera, Superman and Flash travelled to the tomb of Arion, hoping to find the [[Tear of Extinction]], a relic they could use to stop the Triumvirate. Once they arrived, however, the Tear had already been seized by Black Manta.<ref>{{c|Justice League Vol 4 11}}</ref> Fortunately, Mera was able to take the Tear away from Manta, forcing him to retreat.<ref>{{c|Justice League Vol 4 12}}</ref>

The Justice League reunited with Aquaman, who has acquired the [[Trident of Poseidon]], and mounted a final assault on the Triumvirate's stronghold. Although Aquaman was able to make peace the Triumvirate, Manta attempted to use the Triumvirate's last weapon, the Death Kraken, to take over the world. Aquaman was able to stop him, seemingly sacrificing his life in the process, but the League held out hope he had survived and would eventually return to them.<ref>{{c|Aquaman/Justice League: Drowned Earth Vol 1 1}}</ref> Afterwards, a mysterious man calling himself [[William Payton (Prime Earth)|Starman]] had come to the Hall of Justice to warn the Justice League about the fate of the universe,<ref>{{c|Justice League Vol 4 8}}</ref> but his unstable energies began coursing all over Washington D.C. until Wonder Woman restrained him with her lasso. Once his power was stabilized, Starman told Superman and his friends the story of [[Perpetua (Sixth Dimension)|Perpetua]].<ref>{{c|Justice League Vol 4 15}}</ref>

The Justice League, with assistance from Starman and the Green Lantern Corps, traveled to the Source Wall in the hopes of fixing it. To do that, they gathered the surviving Omega Titans and place them along the broken wall. However, their efforts were interrupted by the resurrected Brainiac, who had become a member of Luthor's new Legion of Doom. Brainiac sabotaged the Source Wall's healing process and stole Starman's powers, using them resurrect Perpetua, creator of the Multiverse and the cosmic entity sealed behind the Source Wall. With the Wall destroyed and their mission a failure, the Justice League returned home.<ref>{{c|Justice League Annual Vol 4 1}}</ref>

The Justice League had to ask [[Mister Mxyzptlk (Prime Earth)|Mister Mxyzptlk]] to open a portal to the Sixth Dimension in the hopes of stopping the Legion of Doom and Perpetua. According to Mxyzptlk, Superman was the only one capable of surviving the trip to the Sixth Dimension, so the Man of Steel volunteered to make the trip alone until he could return home and inform the Justice League of his findings. Upon stepping out of the portal, however, Superman found himself trapped in a remote planet surrounded by darkness, with no stars to power him up.<ref>{{c|Justice League Vol 4 19}}</ref> Superman spent an unspecified amount of time trying to escape from the planet until he found himself face-to-face with an older Superman,<ref>{{c|Justice League Vol 4 20}}</ref> who revealed himself to be responsible for Clark's imprisonment. The older Superman easily defeated his younger counterpart in single combat and left him to spend the rest of his life alone in the deserted planet.<ref>{{c|Justice League Vol 4 21}}</ref>.

Even so, Superman refused to give into despair and continued his attempts to escape from the dark planet, using a remote star as his power source. Superman was ultimately able to escape from the planet but the star disappeared, leaving Superman isolated in the middle of dark space.<ref>{{c|Justice League Vol 4 23}}</ref> Fortunately, Superman reached a region of space illuminated by thousands of suns, which revitalized Superman and allowed him to reach the Sixth Dimension, where the Justice League was fighting Superman's jailer, [[Alpheus (Sixth Dimension)|Alpheus]], the World Forger. With his new power, Superman incapacitated Alpheus with a single punch but instead of punishing him for his betrayal, he extended an invitation for Alpheus to join the Justice League and an opportunity to save the Multiverse from Perpetua.<ref name=JL25>{{c|Justice League Vol 4 25}}</ref>

To match Perpetua's army, the Justice League gathered allies from all over the Multiverse and Alpheus suggested they find his brothers, the [[Mar Novu (Sixth Dimension)|Monitor]] and the [[Mobius (Sixth Dimension)|Anti-Monitor]].<ref>{{c|Justice League Vol 4 26}}</ref> Superman partook in the mission to find the Monitor on Nil<ref>{{c|Justice League Vol 4 27}}</ref> but they were unable to find the Anti-Matter on Oward and instead they were forced to escape from an ambush set up by the Legion of Doom.<ref>{{c|Justice League Vol 4 28}}</ref>

The Justice League only had 3 days to stop Perpetua and the Legion of Doom, so Starman formulated a plan to win the oncoming war: to find the missing pieces of the Totality. To that end, the Justice League needed to separate into two teams that would travel to different points in time: the past and the future. Superman, Batman and Wonder Woman would travel to the future and meet [[Kamandi (Prime Earth)|Kamandi]], the last boy on Earth.<ref>{{c|Justice League Vol 4 30}}</ref> He informed Superman's team that the world had been devastated by Brainiac and brought them to the ruins of the Hall of Justice, where the heroes discovered they were trapped inside one of Brainiac's bottles as part of Brainiac's experiments to find to the Totality. Using one of their old ships, the Justice League travelled to a parallel future, where they were attacked by Brainiac's slaves, the [[Justice Legion (DC One Million)|Justice Legion-A]].<ref>{{c|Justice League Vol 4 31}}</ref>

Wonder Woman released the Justice Legion from Brainiac's control and, together, they fought as allies against Brainiac<ref>{{c|Justice League Vol 4 32}}</ref> while Kamandi was tasked with protecting the last fragment of the Totality.<ref>{{c|Justice League Vol 4 33}}</ref> Kamandi brought reinforcements in the battle against Brainiac and gave Superman and his friends enough time to return the fragment of the Totality to their homeworld.<ref>{{c|Justice League Vol 4 34}}</ref> Unfortunately, Perpetua had regained her full power and was preparing to reshape the Multiverse to her will.<ref>{{c|Justice League Vol 4 35}}</ref> The Justice League and their remaining allies prepared for one final battle against Perpetua's army.<ref>{{c|Justice League Vol 4 36}}</ref> Superman and his friends personally confronted Lex, who had become Perpetua's most powerful soldier, amidst the ruins of Washington, D.C.<ref>{{c|Justice League Vol 4 37}}</ref> Martian Manhunter attempted to connect with every single person in the Multiverse in order to stop Perpetua and prove that Justice was superior to Doom, but unfortunately, the Multiverse chose Doom over Justice and the [[Quintessence (Prime Earth)|Quintessence]] quickly brought the Justice League to the Moon and informed them that there was still a way to stop Perpetua, but the Justice League would need to travel to the farthest reaches of the cosmos to find it, a mission the Justice League eagerly accepted.<ref>{{c|Justice League Vol 4 39}}</ref>

====Death Metal====
Unfortunately, the Justice League's mission was a failure and they were forced to return to Earth-0, which was already under attack by an army of Batmen from the Dark Multiverse, led by the [[Bruce Wayne (Earth -22)|Batman Who Laughs]]. Superman himself was defeated by the [[Darkfather (Dark Multiverse)|Darkfather]], imprisoned within the depths of the Sun<ref>{{c|Dark Nights: Death Metal Guidebook Vol 1 1}}</ref> and infected with the Anti-Life Equation. With Superman's defeat, Earth-0 fell under the rule of the Batman Who Laughs and the last remaining superheroes went into hiding.<ref>{{c|Dark Nights: Death Metal Vol 1 1}}</ref>

{{Image|Dark Nights Death Metal Vol 1 5 Exclusive Bulletproof Comics Virgin Variant.jpg|thumb|left|Infected by Anti-Life.}}
Superman was later transferred to New Apokolips, where was tortured by the Darkfather until he was rescued by Batman and Wonder Woman. After defeating Darkfather, Superman helped his friends release an army of superheroes imprisoned within New Apokolips and Wonder Woman enlisted them on all on a mission to defeat Perpetua. For that, Superman and his friends needed to travel to the Dark Multiverse and gather enough power to heal the Multiverse.<ref>{{c|Dark Nights: Death Metal Vol 1 3}}</ref> Superman and his friends infiltrated Castle Bat, home of the Batman Who Laughs, and came across the imprisoned Barbatos, who refused to help the heroes defeat his own treacherous Dark Knight, causing a bored Superman to knock him out. The heroes reached a portal that would take them across different points in the Multiverse's history, each with its own special kind of Crisis energy that could be used to stop the spread of the Dark Multiverse. Superman was sent to the time of the Final Crisis, where he was attacked by Darkseid and a group of Supermen infected by the Anti-Life Equation.<ref>{{c|Dark Nights: Death Metal Trinity Crisis Vol 1 1}}</ref>

Darkseid attempted to turn Superman into another one of his slaves, but Superman was rescued by Wonder Woman, who had convinced [[Kal-El (Earth-Prime)|Superboy-Prime]] to join the Justice League and stop Perpetua. Unfortunately, once they returned to Earth-0, they discovered the Batman Who Laughs had become the Darkest Knight, a cosmic god as powerful as Perpetua, and created his own Multiverse, the [[Last 52 Multiverse]].<ref>{{c|Dark Nights: Death Metal Vol 1 4}}</ref> The Darkest Knight summoned Castle Bat to kill Superman and his friends. Although Lex defeated Castle Bat, Superman was reluctant to trust him until Lex made him realize that only together could they stop the Darkest Knight. Lex convinced a group to build a machine powered by Anti-Crisis energy, the only thing capable of healing the Multiverse. The heroes of Earth-0 raised an army to make a final stand against the Darkest Knight and give Wonder Woman enough time to find the materials to build the Anti-Crisis weapon.<ref>{{c|Dark Nights: Death Metal Vol 1 5}}</ref>

Before the final battle began, however, Superman constructed a time-shifting device that could rewind 60 minutes of time, enough time for Superman to help as many people as he could. The device gave out and Superman returned to the frontlines, but not before spending one last moment of peace with Lois and Jon.<ref>{{c|Dark Nights: Death Metal The Last Stories of the DC Universe Vol 1 1}}</ref> As the Darkest Knight killed Perpetua, Superman and his army prepared to fight the army of the Dark Multiverse, just in time for Wonder Woman, powered by the Anti-Crisis energy, to return to her allies.<ref>{{c|Dark Nights: Death Metal Vol 1 6}}</ref> Surrounded by the Supermen of the Dark Multiverse, Clark attempted to reason with them and make them see the better part of themselves. His words fell on deaf ears, however, and the Man of Steel was forced to fight, even with his powers severely depleted by Anti-Life. Fortunately, Superman's family and a group of his villains, led by Lex were quick to support him in battle.<ref>{{c|Dark Nights: Death Metal The Last 52: War of the Multiverses Vol 1 1}}</ref>

Superman was about to be killed by the [[Last Sun (Last 52 Multiverse)|Last Sun]] until Lex sacrificed his life to save him and destroy the Last Sun, while Wonder Woman destroyed the Darkest Knight, healed the Multiverse and resurrected everyone who had died in the conflict. Cured from Anti-Life, Superman returned home and joined his friends in exploring and protecting the Multiverse they had fought so hard to save.<ref>{{c|Dark Nights: Death Metal Vol 1 7}}</ref>

===[[Infinite Frontier]]===
====Fall of the Golden Age, The Authority & Warworld Saga====
{{Image|Action Comics Vol 1 1036 Textless.jpg|Superman assembles [[Authority (Prime Earth)|The Authority]]|150px|left}}
In an encounter with aliens from the Breach, Superman was exposed to deadly radiation.<ref>{{c|Superman Vol 5 29}}</ref><ref>{{c|Action Comics Vol 1 1029}}</ref> The resulting radiation sickness caused him to age rapidly and his powers to fade, and so he assembled a team to help protect and improve the world if he were to die or be unable to act as Superman anymore.<ref>{{c|Superman and the Authority Vol 1 1}}</ref> He called this team [[Authority (Prime Earth)|The Authority]] and enlisted the aid of two members, [[Enchantress (Prime Earth)|The Enchantress]] and [[Manchester Black (Prime Earth)|Manchester Black]], to conceal and slow the progression of his condition.<ref name=AC_1036>{{c|Action Comics Vol 1 1036}}</ref>

The Authority's first mission became clear when an alien craft carrying refugees from [[Warworld]] crashed to Earth. The refugees were part of a long-lost group of Kryptonians called the [[Phaelosians]] who had been enslaved on Warworld for generations, and Superman swore to free them.<ref>{{c|Action Comics Vol 1 1032}}</ref> The Phaelosians' ship had also been carrying a powerful energy source known as the Genesis Fragment, which [[Atlantis]] laid claim to, heightening tensions with the surface and almost leading to war. Superman confiscated the Fragment and warned both sides against trying to take it, giving the [[Justice League (Prime Earth)|Justice League]] no choice but to expel him.<ref name=AC_1035>{{c|Action Comics Vol 1 1035}}</ref>

Before Superman left for Warworld, Batman asked him for a favor; he had received visions from a wounded and dying [[Fuginauts|Fuginaut]] trapped in the Dark Multiverse.  He asked Superman to bring his new team into the Dark Multiverse and attempt to resolve the situation. Superman agreed, and both entered the Dark Multiverse alongside the Authority in order to combat the [[Empire of Shadows (Earth al Ghul)|Empire of Shadows]].<ref name=BS-AS>{{c|Batman/Superman Authority Special Vol 1 1}}</ref>

{{Image|Action Comics Vol 1 1038 Textless.jpg|Taken captive by the [[Mongul II (Prime Earth)|Mongul-Who-Is]]|150px|right}}
Superman and the Authority journeyed to Warworld, despite warnings that it was almost certainly a trap set by [[Mongul II (Prime Earth)|Mongul]]. Indeed, Mongul was waiting for them and a sorceress working for him reversed the effects of the Enchantress's magic which had been rejuvenating Superman.<ref name=AC_1036/> Superman was forced to fight in the arenas of Warworld, powerless due to the red solar radiation which saturated the world. He asked a veteran Phaelosian gladiator named [[Kryl-Ux (Prime Earth)|Kryl-Ux]] to teach him how to fight without his powers.<ref name=AC_1039>{{c|Action Comics Vol 1 1039}}</ref> Superman became a more adept fighter and managed to inspire some of his fellow slaves, but he received a message from [[Midnighter (Prime Earth)|Midnighter]] saying that he was planning to destroy the star forges which generated the red solar radiation. This would restore Superman's powers but also kill millions of people. Kryl-Ux told him there may be another way and led him to an isolated part of Warworld's catacombs made of the same material as the Genesis fragment. Superman was able to absorb the cosmic energy of the rock and restore his powers.<ref name=AC_1040>{{c|Action Comics Vol 1 1040}}</ref>

In time, Superman earned a reputation as the "Unbloodied Sword", the warrior that spared his opponents instead of killing them,<ref name=AC_1040/> and inspire a true rebellion among the slaves of Warworld. He also became a father figure to two young Phaelosian siblings, [[Osul-Ra (Prime Earth)|Osul]] and [[Otho-Ra (Prime Earth)|Otho-Ra]].<ref name="AC-1042">{{c|Action Comics Vol 1 1042}}</ref> Under the advice of Kryl-Ux, Superman embarked on a quest for the fire of Olgrun, a relic of godly power that could end Mongul's reign over Warworld. Although Superman himself had no use for such power, he still chose to search for it, if only to keep Mongul from using it.<ref>{{c|Action Comics Vol 1 1045}}</ref>

Superman successfully recovered the fire of Olgrun from its resting place, only to find Mongul demanding Superman to surrender the fire, lest Mongul would kill Osul and Otho.<ref>{{c|Action Comics Vol 1 1046}}</ref> Rather than risk letting the fire fall into Mongul's hands, Superman entrusted the fire to Osul, just in time for the rebels of Warworld to modify Warworld's solar energy generators and switch their radiation from red to white. Superman recovered his powers and was cured from the radiation poisoning he got from the Breach, while Mongul died at the hands of Kryl-Ux, who admitted he had used Superman as a pawn in his own scheme to take revenge on Mongul. Although Kryl-Ux escaped, Superman's mission to liberate Warworld was complete, allowing Superman and his allies to triumphantly return to Earth.<ref name=S-WA-V1-1>{{c|Superman: Warworld Apocalypse Vol 1 1}}</ref>

====Dark Crisis====
After spending a day of peace with Lois,<ref name=S-WA-V1-1/> Superman returned to active duty by making an impassionate plea to the United Nations to offer safe haven to Warworld's refugees until they could find a new world to settle upon.<ref>{{c|Action Comics Vol 1 1047}}</ref> He also allowed Osul and Otho to stay over at the Kent apartment and made efforts to protect Osul from Orion, who recognized Osul as the bearer of the fire of Olgrun.<ref>{{c|Action Comics Vol 1 1048}}</ref> Although Orion was forced to retreat, he promised to find a way to remove the fire from Osul, despite Superman's protests.<ref>{{c|Action Comics Vol 1 1049}}</ref>

Superman's reunion with the Justice League, however, was cut short<ref>{{c|Superman: Kal-El Returns Special Vol 1 1}}</ref> when the League was called upon to fight yet another Multiversal threat, this time in the form of [[Kell Mossa (Pre-Crisis)|Pariah]], an ancient scientist whose universe had been destroyed during the Crisis of Infinite Earths. Pariah had channelled the power of the [[Great Darkness]], a primal force that existed before the birth of the Multiverse, to create a Dark Army composed of enslaved supervillains, including Darkseid and Doomsday, and created a machine that would recreate the worlds destroyed during the Crisis on Infinite Earths, a goal that could destroy Earth-0. Although Superman attempted to reason with Pariah, the renegade scientist refused to back down and activated his machine and the Justice League, Superman included, was caught in its energies.<ref>{{c|Justice League Vol 4 75}}</ref>

[[File:Superman Prime Earth 0067.png|thumb|Superman absorbs the power of a pocket dimension to fight Pariah.]]
Pariah's machine did not kill Superman, but rather, it placed him in a pocket dimension that represented Superman's deepest desire: the long-lost desire to see Jon's entire life, from childhood to adulthood. Although Superman was initially happy to live in a world in which he could see his son become a man,<ref>{{c|Dark Crisis: Worlds Without a Justice League - Superman Vol 1 1}}</ref> he came to realize that this world was not real and dedicated himself towards finding a way to escape. He learned to control the illusion's energies and used this power to return to the real world and fight Pariah once again.<ref>{{c|Dark Crisis on Infinite Earths Vol 1 5}}</ref>

Superman and his friends joined Earth's remaining superheroes in fighting Pariah's Dark Army, with Clark personally saving Jon from a still-enslaved Doomsday. The Justice League ultimately sealed Pariah in his own fantasy world, but Deathstroke obtained control of the Great Darkness and attempted to destroy the Multiverse.<ref>{{c|Dark Crisis on Infinite Earths Vol 1 6}}</ref> [[Teth-Adam (Prime Earth)|Black Adam]] transferred his magical powers to Earth's superheroes, allowing them to defeat the Dark Army, while [[Kimiyo Hoshi (Prime Earth)|Doctor Light]] and the Flash Family purified the Multiverse from the Great Darkness' infection. Although the Dark Crisis had ended, the Justice League was disbanded and its individual members chose to go separate ways. Clark and Jon thanked Black Adam for his help during the crisis, acknowledging that, despite their differences, they were in the same side, at least for now.<ref>{{c|Dark Crisis on Infinite Earths Vol 1 7}}</ref>

====Project Blackout and Supercorp====
Soon after the Crisis, Luthor came to the realization that he and Superman should work together. However, he should not be his equal, that Clark should be the outsider alien he always believed him to be. To that end, Luthor initiated [[Project Blackout]], using [[Henry Bendix (Prime Earth)|Henry Bendix's]] technology, Warworld technology and the psychic powers of Manchester Black to erase the memory of Clark and Jon's identities from the world. Superman would see the aftereffects of this as Perry White, being faced with Superman at the Kent Farm, suffered a heart attack as his mind couldn't handle what was seen. Luthor lured Superman to Lexcorp Tower using a special watch based off of [[Jimmy Olsen's Signal Watch]] and revealed what he had done.

Furious, Superman battled Luthor, who ranted that Superman should have told Luthor that he was an outsider like him first and wanted to punish him for telling the world before him. When he revealed that Project Blackout killed Black, Superman becamed enraged over the death of a man he had given a second chance and willingly took it only for Luthor to snuff it out. Defeating Luthor, he took him to jail for Black's murder.

He later met with Batman and Nightwing and learned that, outside of those within the Kent Farm and its barrier when Project Blackout struck, the Justice League, the Titans and their reservists still knew of Clark and Jon's identity thanks to psychic shields by the Martian Manhunter. Batman convinced him to use this as a second chance, as he believed that he, Lois and Jon lost something when he revealed his identity to the world.<ref name=AC-1050>{{C|Action Comics Vol 1 1050}}</ref>

After Lex's defeat, Clark and Lois announced their plans to formally adopt Osul and Otho into the Superman Family, and the media dubbed the young Phaelosians as the "Super-Twins". The reorganized Superman Family's first mission was the defense of the new Steelworks Tower, an institute founded by John and Natasha Irons as a way to properly research and develop new technologies for the benefit of mankind, from a reconstructed and enhanced Metallo.<ref>{{C|Action Comics Vol 1 1051}}</ref> Although the Superman Family foiled Metallo's first assault on Steelworks Tower<ref>{{C|Action Comics Vol 1 1052}}</ref>, Metallo attacked Steelworks again, this time with an army of zombie cyborgs called "Necro-Drones". Metallo ordered his Necro-Drones to self-destruct and injure Supergirl, forcing the Superman Family to help Supergirl while Metallo escaped.

Believing Lex to be responsible for Metallo's rampage, Superman confronted Lex on prison, but Lex claimed no responsibility for Metallo's actions.<ref name=AC-1053>{{C|Action Comics Vol 1 1053}}</ref> Metallo would cause trouble in Metropolis a third time and kidnap the Super-Twins in an attempt to lure Superman into a trap. Fortunately, Superman rescued the Super-Twins and used his Genesis-enhanced powers to defeat Metallo, who claimed that his sister Tracy was in danger and all this time, he had been following the orders of her unseen kidnapper to protect her. Superman agreed to help Metallo save his sister, but nevertheless told him that Metallo would have to pay for all the people he had killed.<ref name=AC-1054>{{c|Action Comics Vol 1 1054}}</ref>

A horde of Necro-Drones attacked A-Town, the apartment complex in Metropolis where the Warworld refugees were housed. While the Superman Family defended A-Town from the Necro-Drones, Superman deduced that Hank Henshaw, the Cyborg Superman, was the real mastermind behind these recent attacks. With the Necro-Drones temporarily defeated, Superman reluctantly activated a copy of the Eradicator to track down Henshaw's latest body to an airport. Superman, his Family and Metallo mounted an assault at Henshaw's base, only to find that Henshaw had transformed Tracy Corben into a cyborg, just like his brother.<ref>{{c|Action Comics Vol 1 1055}}</ref> The Superman Family deactivated Henshaw's Necro-Drones and Metallo destroyed Henshaw, but Tracy is left badly damaged during the battle, but Superman granted the Corben siblings safe asylum at Steelworks Tower.<ref>{{c|Action Comics Vol 1 1056}}</ref>

[[File:Superman Vol 6 3 Textless Nakayama Variant.jpg|thumb|left|Superman and Lex must protect Metropolis together.]]
Shortly before he was arrested, Lex had made Superman the legal owner of LexCorp, which he had renamed "Supercorp" in Superman's honor. Initially, Superman refused to take control of the company formerly owned by his greatest enemy, but [[Mercy Graves (Prime Earth)|Mercy Graves]], Lex's assistant, told him that Lex had threatened to dissolve the company and fire his countless employees unless Superman took control. <ref>{{c|Superman Vol 6 1}}</ref> Superman reluctantly agreed and used Supercorp's resources to protect Metropolis from both the Parasite<ref>{{c|Superman Vol 6 2}}</ref><ref>{{c|Superman Vol 6 3}}</ref> and the Silver Banshee<ref name=S-V6-4>{{c|Superman Vol 6 4}}</ref><ref name=S-V6-5>{{c|Superman Vol 6 5}}</ref>who had been both mutated by a mysterious pair of scientists that, according to Lex, had a vendetta against him. Lex explained that, before Superman's arrival, he tried to be a hero and protect the city's downtrodden masses from Pharm and Graft, two scientists who performing unethical experiments on kidnapped victims and they must have been responsible for the mutations of both Parasite and Banshee.<ref name=S-V6-4/> Later, while trying to find Pharm and Graft, Superman stumbled upon a holographic message left by them in the ruins of Metropolis, warning him about something called "Project Chained".<ref name=S-V6-5/>

In his quest to discover what "Project Chained" was, Superman agreed to use Supercorp's resources and ultimately found this project was an experiment in which Lex attempted to build a special prison for superpowered criminals. Only one prison cell was built, located deep beneath Stryker's Island. Superman and Mercy Graves went there and attempted to free its prisoner, [[Sammy Stryker (Prime Earth)|Sammy Stryker]], only to figure out that Stryker wanted revenge on Lex for imprisoning him in the first place.<ref>{{c|Superman Vol 6 6}}</ref> Superman escaped from the special prison and intercepted Stryker with help from his family<ref>{{c|Superman Vol 6 7}}</ref> and Lex explained his old history with the Chained and that they needed to lock him up again. Superman put on a special suit equipped with Kryptonite gauntlets and kept the Chained occupied long enough for Superboy to put him back into his cell. Unfortunately, Superman was severely exposed to the Kryptonite<ref>{{c|Superman Vol 6 8}}</ref> and had to spend days at Supercorp's medical wing before recovering. Knowing that Pharm and Graft would continue targeting people wronged by Lex, Superman headed back into the field with a special armor, just in time to help [[Marilyn Moonlight (Prime Earth)|Marilyn Moonlight]] in a fight against Graft, who immediately teleported them into the days of the Wild West.<ref>{{c|Superman Vol 6 9}}</ref>

[[File:Superman Vol 6 10 Textless.jpg|thumb|right|Superman in the Wild West.]]
Superman and Marilyn spent days trapped in the Wild West, looking for a way to return home, and confronted Terra-Man, who had equipped they could use to return to present. After defeating Terra-Man and returning home, Superman went to the Supercorp tower to see if his employees were safe, only to be ambushed by Pharm and Graft, who had formed a new Lex Luthor Revenge Squad and wanted Superman to join them.<ref>{{c|Superman Vol 6 10}}</ref> Graft exposed Superman to red Kryptonite and attempted to make him kill Lex, only for Lex to bring him back to normal by exposing him to green Kryptonite. Unfortunately, Lex's mother, [[Leticia Luthor (Prime Earth)|Leticia]], had conspired with the Revenge Squad to release a cloud of Kryptonite gas all over Metropolis in an attempt to destroy his son's legacy.<ref name=S-V6-11>{{c|Superman Vol 6 11}}</ref> Lex defeated his mother while Superman neutralized the Kryptonite cloud, saving the city, but Superman couldn't help but wonder why did the Revenge Squad wait for this particular moment to restart their vendetta against Lex.<ref>{{c|Superman Vol 6 12}}</ref> 

====The Empire of Shadows and the Planet of Bizarros====
A radical humanocentric movement, the [[Blue Earth Movement|Blue Earth]], started to gain popularity in Metropolis and its leader, Norah Stone, offered an exclusive interview to the ''Daily Planet''. Clark was put in charge of the interview and, although he deduced that Stone was lying, she was able to frame the Blue Earth's mission as the preservation of humanity. During and after the interview, Clark grew steadily weaker and was attacked by Stone's bodyguard, Dorian, who had somehow drained Clark of his powers.<ref>{{c|Action Comics Vol 1 1058}}</ref> Luckily, Clark used Lex's old anti-Kryptonians weapons and Dorian's own inexperience with Kryptonian powers to defeat him and discovered that he had used a mysterious green bracelet to regain his powers. Still, Clark's powers would take time to fully recharge, so he asked John Henry to build a special armor for combat.<ref>{{c|Action Comics Vol 1 1058}}</ref>

With his new armor, Superman easily defeated a gang of Blue Earthers that somehow gained Kryptonian powers and were hurting criminals. Superman's powers were slowly restored, and he deduced that Stone was giving powers to her followers to demonize aliens and strengthen her political stance. He then remembered his old adventure with Batman and the Authority into the Dark Multiverse<ref name=BS-AS/> and realized that Stone was actually [[Janan al Ghul (Earth al Ghul)|Janan al Ghul]], and she had just kidnaped Otho.<ref>{{c|Action Comics Vol 1 1059}}</ref> To find Otho, Superman asked [[John Constantine (Prime Earth)|John Constantine]]'s help and they invaded Blue Earth's headquarters and liberated [[Etrigan (Prime Earth)|Etrigan]] and [[Raphael Arce (Prime Earth)|Bloodwynd]] from their prison, and together, they travelled to [[Earth al Ghul]] to save Otho and stop Janan.<ref>{{c|Action Comics Vol 1 1060}}</ref>

Upon their arrival, Superman and his allies attacked the al Ghul's citadel, just as Janan was preparing her invasion of Earth-0. Janan had also brainwashed Otho into attacking her adoptive father, but Superman's love for her daughter allowed her to break free from the brainwashing and Janan was forced to begin her invasion prematurely. Superman and his allies returned home, just in time for the Superman Family and the Blue Earthers to join forces and defend their world together. After a fierce battle, Janan is defeated and Etrigan throws her soul into Hell. <ref>{{c|Action Comics 2023 Annual Vol 1 1}}</ref> Unfortunately, peace did not last for long, as Bizarro returned to wreak havoc on Metropolis yet again. Bizarro had stolen multiple magical artifacts from across the multiverse and began casting a spell that would have severe consequences on the city and its people. Superman believed Bizarro had lost control of the spell and tried to save him, but Bizarro completed the spell and disappeared in a mystical explosion. Suddenly, everyone in the city began transforming into Bizarros, with the original Bizarro taking control of Superman's mind and body. <ref>{{c|Action Comics Vol 1 1061}}</ref>

[[File:Action Comics Vol 1 1063 Textless.jpg|thumb|left|Battle for the soul of Metropolis.]]
Through sheer willpower, Superman fought Bizarro's mental domination and tried to find a way to undo his spell, all the while trying to prevent the citizens of Metropolis from hurting each other. The superhero community isolated Metropolis from the rest of the world and Superman elected to stay behind and help the people in whatever way he could, but despite his best efforts, the quarantine failed, and Bizarro's spell ultimately spread across the world. Superman believed himself to be the only person unaffected by the spell, so he decided to request assistance from the only sane person left on Earth: the Joker.<ref>{{c|Action Comics Vol 1 1062}}</ref> Using Kryptonian meditation techniques and guidance from the Bizarro Joker, Superman travelled deep into his mind to confront Bizarro, who was still trying to usurp control of Superman's body. After a long mental battle in which Superman died repeatedly, the Man of Steel ultimately gained Bizarro's magical knowledge and used it to undo his spell, even if that meant turning the Joker back into a ruthless murderer. Although the crisis was over, a small piece of Bizarro remained in Superman's mind, but Superman promised to himself to ignore and just be the best version of himself.<ref>{{c|Action Comics Vol 1 1063}}</ref>

====The House of Brainiac and Absolute Power====
Just as Clark and Lois were enjoying a day off, Brainiac would an unleash an army of Czarnians and robots upon Metropolis and orchestrate the kidnapping of every single metahuman in the city, including members of the Superman Family. Superman faced off against the Czarnian general [[Chacal (Prime Earth)|Chacal]] and demands to be taken to Brainiac, but Lex offered himself in Superman's place, as Brainiac had kidnapped Lex's daughter, Lena, and needed Superman to save their people. After the attack, Superman decided that the only way to find Brainiac would be to find Lobo.<ref>{{c|Action Comics Vol 1 1064}}</ref> He tracked Lobo down to Coast City and angrily demanded to learn about his connection to Brainiac but Lobo replied that he had no idea what Superman was talking about, until Superman showed him footage of Brainiac's attack and Lobo was surprised to learn that Chacal was involved and decided to help Superman find Chacal, whom Lobo considered a personal idol. The United Planets had established a blockade around the Earth, and to get past it, Supercorp provided Superman and Lobo with spaceworthy motorcycles. Although Superman and Lois successfully bypassed the blockade, they are intercepted by Chacal's forces<ref>{{c|Superman Vol 6 13}}</ref>

Chacal and his thugs distracted Superman and Lobo enough for Brainiac's ship to escape from Superman's sight, so Lobo took the Man of Steel to Hardcore Stations, a place where they could find information about Brainiac. Unfortunately, Chacal tracked them down and convinced Lobo to betray Superman in exchange for an opportunity to replace him as leader of his gang.<ref>{{c|Action Comics Vol 1 1065}}</ref> With help from Chacal and his men, Lobo captured Superman and took to Brainiac, who was already waiting for him. Brainiac connected Superman and other metahumans to a special machine that power up his latest creation: a [[Brainiac Queen (Prime Earth)|Queen]] that could absorb the powers of other metahumans. Upon her activation, the Queen was consumed by a destructive appetite that could her to kill Chacal and his men.<ref>{{c|Superman Vol 6 14}}</ref>

Angry at Brainiac's betrayal, Lobo released Superman, who told his allies to protect Lena, who had been used by Brainiac as a living processor for his ship. Using his tactile telekinesis, Superboy disabled the Queen while Superman and his allies labored to save the cities bottled within Brainiac's ship. Lex came up with a plan to defeat Brainiac but it would involve him and Lena to get inside Brainiac's network and shut it down from within while Superman and his Family fought Brainiac and his forces directly.<ref>{{c|Action Comics Vol 1 1066}}</ref> Lex and Lena managed to delete Brainiac's entire database but his ship sustained heavy damage during the battle. Brainiac, having lost his entire network and consumed by irrational emotion, begged Superman to save the last city of Colu, which he had preserved before the [[Omega Titans]] destroyed the planet. Superman agreed and forcibly separated the Coluan city from Brainiac's ship before it exploded. The survivors from Brainiac's countless attacks settled on an artificial planet called "Colu 2.0" to rebuild their civilization, while Superman and his family returned. Unfortunately, exposure to Brainiac's databse had left Lex with amnesia.<ref>{{c|Superman Vol 6 15}}</ref>

[[File:Superman Prime Earth 0030.jpeg|thumb|left|Without his powers, Superman takes a bullet.]]
Superman and his allies would not have time to rest on their laurels, however, as Amanda Waller had unleashed an group of robots, [[Task Force VII (Prime Earth)|Task Force VII]], to steal the powers of every metahuman on Earth. One of those robots, the [[Last Son (Prime Earth)|Last Son]], stole the powers of not just Superman, but also those of Conner and Jon. Superman fell victim to Last Son when he was busy stopping a robbery and thus took a bullet to the chest while Last Son kidnapped Jon.<ref>{{c|Absolute Power Vol 1 1}}</ref> Superman was rushed to the Fortress of Solitude, which had become a makeshift refuge to any victim from Task Force VII, and was subjected to a surgery by [[Michael Holt (Prime Earth)|Mr. Terrific]] and [[Pieter Cross (Prime Earth)|Dr. Mid-Nite]]. Although his life was saved, the number of depowered superhumans was growing and Zatanna, who had been taken to the Fortress by Mercy, suggested a plan that could help them beat Waller, but it involve crossing into the Dark Roads, a realm of magic.<ref>{{c|Superman Vol 6 16}}</ref>

At the Fortress, Nightwing assumed leadership of the depowered superheroes, while Superman and Batman would outfit their allies with special suits that would replicate their lost powers. Unfortunately, the Fortress would be hit by the Brainiac Queen, reconstructed by Waller to serve her cause. During the battle, Superman would be treated to a terrible sight: Jon had been brainwashed and turned into a slave for Waller. The Man of Steel tried to help his son, but he was ultimately forced to order a retreat and activate the Fortress' self-destruct protocol. Superman was almost captured by the Queen and it was only through [[Nia Nal (Prime Earth)|Dreamer]]'s sacrifice that Superman escaped into the Phantom Zone and regroup with his friends.<ref>{{c|Absolute Power Vol 1 2}}</ref> The world's superheroes set up a new base of operations in Themyscira, while Superman and Zatanna travelled to the Justice League Dark's old headquarters and found a passage to the Parliament of Trees. During their journey, they were attacked by Waller's forces until they found a portal to the Oblivion Bar, where the demon lord [[Neron (Prime Earth)|Neron]] was waiting for them. He said he could give them the Map of Mordru, an object that could help them navigate the Dark Roads, but they have to pay a high price for it.<ref>{{c|Superman Vol 6 17}}</ref>

Instead of asking for Superman's soul, Neron asked Superman, a living embodiment of truth, to lie, a clear violation of his principles and ideals; and Superman accepted. Zatanna was concerned over Superman's decision, but the Man of Steel believed the importance of their mission was worth the stain on his conscience. With the map, Superman and Zatanna gained access to the Dark Roads but the map started draining Zatanna's life force in the process. Upon reaching the [[Monitor's Satellite|House of Heroes]], they found it destroyed and only [[Kimiyo Hoshi (Prime Earth)|Dr. Light]] present. Suddenly, they were attacked by [[Konfusion (Earth 3)|Konfusion]], one of the [[Bureau of Sovereignty]]'s agents, but Superman and his friends escaped from the Bleed and returned home to reunite with Lois and Mercy, who were waiting for Superman at the ruins of the Fortress. Lois and Mercy decided to return to Metropolis and publish the truth about Waller's actions, while Superman travelled to Themyscira, which had become a temporary base of operations for the superhuamn resistance, to prepare for a final attack on Waller'a base.<ref>{{c|Superman Vol 6 18}}</ref> The superheroes endured an attack from the Brainiac Queen and released Jon from her control, much to Superman's relief.<ref>{{c|Absolute Power Vol 1 3}}</ref>

The united heroes began their assault on Waller's base of operations, located deep in Gamorra, and faced heavy opposition from the Bureau of Sovereignty and Task Force VII. Luckily, the tide turned in the heroes' favor when Green Arrow, having revealed himself to be a mole within the Bureau, shot a special device invented by Batman at Waller's robots, causing them to lose their powers and the heroes, Superman included, to regain them. With Task Force VII destroyed and Waller detained, Superman helpled Hal Jordan and the Flashes stop a multiversal army of villains from invading their universe, but this was not a clean victory for the heroes. As the Fortress of Solitude underwent extensive reconstructions, Superman, Batman and Wonder Woman reprimanded Green Arrow for his undercover actions, while Green Arrow countered that Waller would not have given them trouble in the first place had the Justice League been there to stop her, a sentiment Superman and his friends agreed with, and so, efforts to rebuild and expand the League began.<ref name=AP-4>{{c|Absolute Power Vol 1 4}}</ref>

===[[All In]]===
====Journey into the Phantom Zone====
As the world struggled to recover from Waller's attempt to seize power, Superman and Supergirl were attacked by three Kryptonians from the Phantom Zone that had been forcibly combined into a single, hideous creature. With help from Jimmy and Supergirl, Superman removed the creature's powers with gold Kryptonite and took it to Kandor, where a scientist explained that Kandor had lost its surveillance equipment on the Phantom Zone after the Brainiac Queen had attacked the Fortress. Superman theorized that the Zone was undergoing some kind of change and decided to investigate. He took with him special bracelets that would keep his powers replenished in the sunless Zone and left his family to defend Earth in his absence.<ref>{{c|Action Comics Vol 1 1070}}</ref> As Superman materialized within the Zone, he was attacked by a band of criminals that still held a grudge against Jor-El for imprisoning them. He quickly captured the criminals and forced them to take him to their boss, until they were ambushed by the Val-Kon, the ship of Xa-Du. When Superman was brought onboard, however, he was surprised to see that it was not Xa-Du who was in charge of the ship, but his old friend Mon-El.<ref>{{c|Action Comics Vol 1 1071}}</ref>

[[File:Action Comics Vol 1 1073 Textless.jpg|thumb|right|Brothers reunited.]]
Mon-El, having imprisoned Xa-Du and taken control of his ship in Superman's absence, explained that the Zone was becoming a physical realm thanks to the actons of an entity known as Aethyr, who had promised to release the Zone's prisoners from their eternal torment. Aethyr created physical realms within the Zone and made the Zoners corporeal again. The violent criminals quickly returned to their old ways once they were able to hurt each other, and a disgusted Aethyr enslaved them and appointed Xa-Du as his warden. When the Val-Kon passed an immaterial region in the Zone, Xa-Du took this opportunity to escape and reunite with Aethyr.<ref>{{c|Action Comics Vol 1 1072}}</ref>

To stop Aethyr, Superman and the Val-Kon's crew travelled to Aethyropolis, the center of Aethyr's power. There, they were attacked by Aethyr's army and Xa-Du destroyed the device Superman was going to use to return to Earth. Superman was ambushed by Aethyr, who was displeased by Xa-Du's failure to control the Phantom Zone and threw both Superman and Xa-Du to the eye of the Phantom Zone, a place in which multiple points of time and space were intertwined. Superman attempted to escape back to Earth by crossing one of the countless cracks in the void, but instead of reaching Earth, he was thrown into Krypton's past.<ref>{{c|Action Comics Vol 1 1073}}</ref> While exploring the city of Kryptonopolis, Superman deduced that he had travelled back to a time shortly before Krypton's destruction and resolved to find a way home. To do that, he hid his identity under the alias "Klar Ken" and went to the House of El for help in recharging his solar batteries. Jor-El invited Superman to his lab and revealed his theory that Krypton would soon explode, but suddenly, representative from the Science Council called Ro-Zan burst into the lab and arrested Jor-El for developing technology without the Council's permission. Superman attempted to defend his father but was knocked out and arrested, as well.<ref>{{c|Action Comics Vol 1 1074}}</ref>

After their arrest, Superman, Jor-El and Lara were called to stand trial before the Council for the charge of hoarding special technology for themselves. Because of his lifetime of achievements, the Council has agreed to let Jor-El and his allies walk free, but his latest invention, the Phantom Zone projector, would be left under the Council's custody, much to Jor-El's protests. This development shocked Superman as, all this time, he had believed that his father had proposed the usage of the Phantom Zone as a form of rehabilitation for dangerous criminals. Upon leaving the Council's chambers, Jor-El revealed that he and Lara secretly recovered Superman's solar bracelets. Jor-El took Superman to the home of his brother Zor-El, where they hoped they could repair and recharge the bracelets.<ref>{{c|Action Comics Vol 1 1075}}</ref>

Zor-El told Superman that he could recharge his solar batteries with his photonic converter, but Ro-Zan has already confiscated it. When a groundquake hit Argo City, Superman told Zor-El that his brother's warnings about Krypton's impending doom might be correct. Zor-El angrily dismissed that theory, for he believed that his brother was disgracing the El name. He threw Superman and the Els out of his house and told them never to come again. With no other options, Superman was forced to sneak into the Science Council's building and try to find Zor's photonic converter so he could recharge his solar bracelets. With his powers fully recharged, Superman returned to the Phantom Zone using the same portal that took him to Krypton in the first place, but upon his arrival, he found that the Zone was starting to collapse.<ref>{{c|Action Comics Vol 1 1076}}</ref> The Zone was becoming increasingly unstable due to the presence solid matter, and Superman reunited with Mon-El, who revealed that Aethyr had been using Earth's sun as a power source to create solid matter within the Zone. Superman begged Aethyr to stop his plans to reshape the Zone, but Aethyr refused to listen and instead used more solar energy to increase his size until he was crushed under his own mass. All that remained of Aethyr was a sphere he had been using to convert energy into matter and Mon-El used it to create a portal that led Superman back to Earth. <ref>{{c|Action Comics Vol 1 1077}}</ref>

Superman returned to Earth just in time to save his fellow members of the Super-Family from a fall from orbit after they had exhausted their solar powers in a conflict with the villainous Grax. At the Fortress of Solitude, Mr. Terrific explained that the Sun's decay could be halted if Superman flew deep into the Sun and destroyed the device Aethyr had planted there to convert solar energy into matter within the Phantom Zone. He also warned that destroying the device would likely have catastrophic consequences for the Zone and its prisoners, which is why Superman gave the Phantom Zone projector to Jon and Supergirl and instructed them to travel to a distant planet with a red sun and evacuate the prisoners there. Such plan, however, was met with opposition from Bry-Zan, a high-ranking elected official from Kandor and daughter of Ro-Zan. She warned that releasing prisoners from the Zone without authorization from Kandor would result in Superman's permanent banishment from Kandor, but Superman chose to stick to the plan and save as many as he could. Outfitted with a special armor constructed by his friends, Superman flew into the Sun and successfully destroyed Aethyr's device. Although the Sun was revitalized and Superman's powers were replenished, the Man of Steel was horrified to discover that the Zone's prisoners had escaped and gained powers, as well.<ref>{{c|Action Comics Vol 1 1078}}</ref>

The Kryptonian criminals severely injured Superman and went on to attack the Earth, while Mon-El rescued Superman and took him to a lead mine in India where he could recover. Superman woke up after two days of rest, and Jimmy and Mon-El informed him about what had happened when he was unconscious. Once he joined the defense of Earth, Superman tried to find his Gold Kryptonite isotope so he could use it to depower the Phantom Zone escapees, only to discover that someone else had taken it. Although the Super-Family and the Justice League successfully apprehended a huge portion of the Zoners, Batman informed Superman that some of the Kryptonians had already spread out into outer space and were attacking foreign planets.<ref>{{c|Action Comics Vol 1 1079}}</ref> The heroes of Earth mobilized teams across the galaxy to capture the Zoners as quickly as possible, with Superman working alongside Mon-El to defeat Ras-Krom on Grenda. They later went to Braal to intercept Jax-Ur, who revealed that he had stolen the Gold Kryptonite hidden by Superman, but luckily, Mon-El trapped Jax-Ur back in the Zone before he could use it.<ref>{{c|Action Comics Vol 1 1080}}</ref>

Mon-El's special equipment, which allowed him to interact with the material world, became damaged and he was forced to return to the Zone, leaving Superman and his friends to continue the mission on their own. To capture the Zoners much more quickly, Superman and his friends built a large Phantom Zone projector and set up a trap to lure them into the projector, with Martian Manhunter assuming the form of Xa-Du to attract the escapees into the trap. The plan worked, and the escapees, including Xa-Du, were sent back into the Zone again. Supergirl also convinced the council of Kandor to allow Zoners for the possibility of parole. Mon-El was selected to be the Zone's "warden" and Superman reaffirmed his promise to find a way to cure his friend and release him from the Zone.<ref>{{c|Action Comics Vol 1 1080}}</ref>

====Birth of Superwoman and Return of the Justice League====
[[File:Justice League Unlimited Vol 2 1 Textless.jpg|thumb|left|A new beginning for the Justice League.]]
Superman invited as many heroes as he could to the reformed Justice League and was present at the opening ceremony of the new Watchtower, an event interrupted by a sudden assault from Darkseid, who had absorbed the Spectre to increase his power.Zatanna enchanted Superman with a spell that not only released the Spectre from Darkseid's control, but seemingly destroyeed the Dark Lord of Apokolips, as well. Darkseid's apparent death opened a gateway to the mysterious Elseworld, Earth-Alpha, and Superman declared that the new League's first official mission had begun. Booster Gold volunteered to investigate Earth-Alpha, but the League lost contact with him as soon as he stepped into the gateway.<ref name=DC-AIS>{{c|DC All In Special Vol 1 1}}</ref> Unfortunately, the new League couldn't spare the time to locate Booster, for it had to answer the threat of the mysterious terrorist organization Inferno, which had unleashed an army of machines to attack a village in South Africa.<ref>{{c|Justice League Unlimited Vol 2 1}}</ref>

Although Waller was imprisoned and her forces had been detained, some of the superpowers they had stolen had failed to return to their original owners. Some superhumans had switched powers, while other powers have been spread out across the entire globe and found new owners.<ref name=AP-4/> One of those new superhumans was Lois, who had acquired powers similar to those of Superman, and she used them to go on many adventures with her husband. Superman and the new Superwoman were ambushed by Doomsday, who had escaped imprisonment from Hell. To make matters worse, Superman was visited by a future Doomsday who had become the latest entity to bear the mantle of the Time Trapper.<ref>{{c|Superman Vol 6 19}}</ref> The Time Trapper wanted Superman's help to evolve to a newer, more powerful form and stated that everyone Superman knew and loved would be destroyed by an unknown enemy. Superman refused to help the Time Trapper, who simply departed and said that Superman would help him sooner or later. Present Doomsday continued his attack on Metropolis and Supercorp attempted to stop him by having Parasite absorb his power but Doomsday ripped Parasite in half, and Superman was forced to beat Doomsday into unconsciousness so that Supercorp could capture him.<ref>{{c|Superman Vol 6 20}}</ref>

After Doomsday's capture, Superman took Superwoman on a date, during which they performed a few superhero activities together. He even confessed that the injection of red Kryptonite he got from the Lex Luthor Revenge Squad<ref name=S-V6-11/> was still affecting him. Their date ended with the arrival of the Aftermath, a collective of alien survivors whose planets had been destroyed by Doomsday, above the skies of Metropolis. The Aftermath demanded that Doomsday be handed over to them, lest they would destroy the Earth.<ref>{{c|Superman Vol 6 21}}</ref>

| Powers                  =
* {{Alien Physiology|Kryptonian}}: Superman's Kryptonian biology is able to store solar energy actively within his bio-cellular matrix as an energy pattern that is linked to his body's electromagnetic field. This process powers most of his electromagnetic capabilities such as flight, heat-vision and other "sight"-based abilities, while also supplementing his physical abilities to immense superhuman levels.<ref name=SMU8/> Aside from his incredible power, Superman is integral to the balance of the universe. [[Aztar (Prime Earth)|The Spectre]] explained that Superman embodies the primal forces of Life and Hope, which push reality outwards to expand. This force is counterbalanced by Constraint, which holds the expansion back, and is embodied by [[Darkseid (Prime Earth)|Darkseid]]. Together, they maintain the balance of the universe.<ref name=DC-AIS/>
** {{Energy Absorption|Solar Radiation Absorption}}: Due to his Kryptonian biology, Superman's body absorbs solar energy and converts it into power, causing him to pull in 140 gigawatts of solar energy at any given moment.<ref>{{c| Superman Unchained Vol 1 2}}</ref>. Stars of different frequencies give him different abilities. Earth's yellow sun fuels all of his powers, while the wavelength of his home solar system's Red Sun enables his body to function on an identical level of an ordinary human. Every time Superman uses any of his superhuman abilities, his body expends absorbed sunlight. By flying into the sun, Superman is able to recharge his energy and temporarily boost his powers. After being exposed to [[Warworld]]'s [[Genesis Fragment|Genesis]]-powered "White" Sun, Superman's existing powers were greatly enhanced and granted new abilities. After the White Sun exposure, Superman claimed that concepts like weight, distance, temperature and even space-time have largely lost their meaning for him.<ref name=AC-1050/>
*** {{Energy Absorption}}: Superman's solar receptors have allowed him to absorb different types of energy in certain instances. He has absorbed  cosmic energy from [[William Payton (Prime Earth)|Starman]], gained from the [[Totality (object)|Totality]] itself.<ref>{{c|Justice League Vol 4 15}}</ref> This boosted his powers to the point that he felt like he could crack the Earth in half with a footstep, if he wasn't careful.<ref>{{c|Justice League Vol 4 15}}</ref> Superman also displayed the ability to absorb the radiation from an atomic explosion, however this left him severely weak.<ref>{{c|Superman/Wonder Woman Vol 1 7}}</ref>
** {{Superhuman Strength}}: Superman possesses near limitless strength, though the exact limits of Superman's strength is unknown. Superman is one the most physically strong beings in the multiverse. His immense willpower allows him to push beyond his physical limits when necessary, allowing Superman to exert incredible feats of strength. He was strong enough to knock out a restrained [[Barbatos (Dark Multiverse)|Barbatos,]]<ref name="DKM-4"/> with a just a punch.<ref>{{c|Dark Nights: Death Metal Trinity Crisis Vol 1 1}}</ref> When Superman fought against [[Rogol Zaar (Prime Earth)|Rogol Zaar]] inside the [[Phantom Zone|Phantom Zone]], their fight made the realm tremble and shake, under the weight of their battle.<ref>{{c|Superman Vol 5 6}}</ref> His powers were further augmented by Warworld's Genesis-powered "white" sun, making him strong enough to fight two [[New Gods]] simultaneously.<ref name=AC-1049>{{c|Action Comics Vol 1 1049}}</ref> After being supercharged by multiple suns, Superman was able to punch the [[Alpheus (Sixth Dimension)|World Forger]] hard enough to create a continent-sized hole, and crack a tear in the [[Sixth Dimension]].<ref name=JL25/>
*** {{Sound Manipulation|Shock Wave Projection}}: Superman is strong enough to create powerful sonic booms with just the clap of his hands. These sonic claps are powerful enough to be heard across states.<ref>{{c|Justice League/Aquaman: Drowned Earth Vol 1 1}}</ref>
** {{Superhuman Speed}}: Superman is able to move at incredible speed by sheer force of will. This extends to his perceptions and allows for feats such as catching bullets in mid flight as well as covering vast distances in little or no time. Superman was even fast enough to keep up with [[Flash (Wally West)|Wally West]] on foot, though he is still not as fast.<ref>{{c|Titans Vol 3 7}}</ref> He can move so fast that time slows from his perspective, long enough to have an entire conversation in just a few seconds.<ref>{{c|Superman Vol 5 11}}</ref> Superman is fast enough to essentially be in two places at once, as he was capable of fighting [[Darkseid (Prime Earth)|Darkseid]] and saving the lives of others simultaneously, which perplexed even Darkseid himself.<ref>{{c|Superman: Man of Tomorrow Vol 1 16 (Digital)}}</ref> After being supercharged by multiple suns, Superman was able to move at a speed described as beyond physics and imagination, and far past the speed of light.<ref name=JL25/> Superman is fast enough to jump into the [[timestream]] and travel throughout it, with the aid of the [[Cosmic Treadmill]], something only the [[Flash Family]] was thought fast enough to do.<ref name=AC992/>
*** {{Molecular Acceleration}}: Superman can vibrate his molecular structure to achieve various effects.<ref>{{c|Batman/Superman: World's Finest Vol 1 1}}</ref> By vibrating fast enough, Superman can enter the "space between spaces", a vibrational frequency that nobody can perceive, thus giving him and [[Dru-Zod (Prime Earth)|General Zod]] the chance to talk privately.<ref>{{c|Superman Vol 5 12}}</ref>
**** {{Phasing}}: By vibrating his molecules on an atomic level, and by doing so at just the right frequency, Superman can phase through objects such as walls without damaging them.<ref name=AC-1053/>
**** {{Intangibility}}: Superman can vibrate his molecular structure to an intangible state, allowing objects to pass through him, including energy blasts.<ref name=AC-1049/>
*** {{Aerokinesis|Vortex Creations}}: Superman can use his speed to manipulate airflow for various purposes. He was able to use his flight speed to funnel a toxic chemical out of the atmosphere and into space.<ref name=AC990/> By flying around a group of enemies at super speed, Superman can create a vortex around them to draw away their oxygen and choke them out.<ref>{{c|Batman/Superman: World's Finest Vol 1 33}}</ref>
** {{Superhuman Agility}}: Superman's agility, balance, and bodily coordination are enhanced to superhuman levels. This allows him to easily make sharp turns and easily maneuver while flying and moving at superhuman speed.
** {{Superhuman Reflexes}}: Superman possesses reflexes far greater than any normal human being. This allows him to easily react to danger and events, and perceive the world in slow motion.<ref name=AC_1000/>
** {{Superhuman Stamina}}: Superman is able to maintain continuous strenuous physical action for an indefinite period of time. This is based on his body converting yellow solar radiation directly to energy but is limited by physiological and psychological needs to eat, drink and sleep.
** {{Enhanced Senses|Superhuman Senses}}: All of Superman's senses are enhanced to superhuman levels, allowing him to perceive the world in far greater detail than the average human.<ref name=AC_1000/> By combining his super-hearing with the sound waves from his sonic-claps, he can simulate a rough form of echolocation.<ref>{{c|Action Comics Vol 1 1071}}</ref>
*** {{Enhanced Vision|Super-Vision}}: Superman's vision processes the entire electromagnetic spectrum as well as allowing vast control over selective perception and focus. This umbrella ability includes the following:
**** '''Electromagnetic Spectrum Vision''': Superman can see well into most of the electromagnetic spectrum. He can see and identify radio and television signals as well as all other broadcast or transmitted frequencies. Using this ability, he can avoid detection by radar or satellite monitoring methods. This also allows him to see the aura generated by living things.
**** '''[[Enhanced Vision#Infrared Vision|Infrared Vision]]''': Superman can see with better acuity in darkness, and to a degree in total darkness.
**** '''[[Microscopic Vision]]''': This is the ability to see extremely small objects and images down to the atomic level.
**** {{Radiation Production}}: Superman can emit various forms of radiation through his eyes. He once used this ability to emit microwave radiation to fry the electronics in a room.<ref name=AC2/>
**** '''[[Enhanced Vision#Telescopic Vision|Telescopic Vision]]''': This is the ability to see something at a great distance, without violating the laws of physics. Though limited, the exact extent of the ability is undetermined. In function, it is similar to the zoom lens on a camera.
**** '''[[X-Ray Vision]]''': By focusing his vision, Superman can see through any volume of matter, except lead. He can see things behind a solid, opaque object as if it were not there and can focus this ability to "peel back" layers of an object, allowing hidden image or inner workings to be observed. Superman's enemies have attempted to use lead to hide things from his vision, mistakenly believing that anything with that material is somehow invisible to him. On the contrary, lead's opaqueness to his X-ray vision makes the material easy to spot with a wide-angle scan and allows Superman to narrow his search to those opaque objects.<ref>{{c|Action Comics Vol 1 1024}}</ref> He can even use intensify his X-Ray vision to the extent that he could use it at a Gamma level to short out a Space Station's electrical engine.<ref>{{c|Superman Unchained Vol 1 1}}</ref>
*** {{Enhanced Hearing|Super-Hearing}}: Superman's hearing is sensitive enough to hear any sound at any volume or pitch. With skill and concentration, he can block out ambient sounds to focus on a specific source or frequency.
*** '''Superhuman Taste''': Superman's sense of taste is enhanced to levels far greater than any normal human. He is capable of singling out every specific ingredient in food by tasting it.<ref name=SM39/>
** {{Accelerated Healing}}: Superman's bodily tissue can repair itself fast enough to be visibly noticeable, allowing him to heal from injuries at a highly accelerated rate.<ref name=AC2/> The time it takes for him to heal depends on the level of injury. When under direct sunlight, Superman can heal almost instantly.<ref name=AC2/>
*** {{Toxic Immunity}}: Superman is resistant to any toxins and diseases due to his superhuman immune system.<ref>{{C|Action Comics Vol 2 20}}</ref>
** {{Enhanced Intellect|Enhanced Mental Process}}: Superman's mental capacity is enhanced to levels far greater than any normal human, allowing him to take in and process information at an accelerated rate. This allows Superman to read through large amounts of information at super-speed and retain it.<ref name ="AC12"/><ref name ="AC14"/> His mind is able to process information at such speeds that he can perceive femtoseconds and attoseconds.<ref name=AC_1000/>
** {{Flight}}: Superman is able to manipulate graviton particles to defy the forces of gravity and achieve flight. This ranges from hovering to moving in any posture, in any direction. He is capable of flying at above supersonic and hypersonic speeds.<ref name=AC_1000/>
** {{Thermal-Blast|Heat Vision}}: Superman can fire beams of intense heat at a target by looking at it. He can vary the heat and area affected. After receiving brief training from [[Wraith (Prime Earth)|Wraith]] on how to use his heat vision for more combative purposes, Superman learned to rapid fire his heat vision at maximum heat, by blinking at super speed.<ref>{{c|Superman Unchained Vol 1 4}}</ref>  
*** {{Energy Projection|Solar Flare}}: Superman's last resort power that he'd discovered through his heat vision, was the ability to immediately expel all of his stored solar energy in one gigantic burst of explosive force. This Solar Flare uses all the stored energy within his cells at once, allowing for an all-out attack.<ref name=SM-3-38/> He cannot use it more than once or twice however, as it drains him of all his powers until he can recharge his cells with solar energy. Give or take 24 hours, in that expanse of time, he is rendered virtually human and vulnerable.<ref>{{c|Batman/Superman Annual Vol 1 2}}</ref><ref>{{c|Superman/Wonder Woman Annual Vol 1 2}}</ref><ref name=SM39>{{c|Superman Vol 3 39}}</ref> Superman eventually learned to keep his flares smaller and more contained, so that it doesn't take a toll on his powers.<ref name=SM-3-41/> Cyborg stated that this Solar Flare has the effects of an EMP.<ref>{{c|Superman Vol 3 40}}</ref>
** {{Immortality}}: Superman is unable to die from old-age and can seemingly live forever. The [[Time Trapper]] stated that Superman would outlive everything in the universe.<ref>{{c|Superman Vol 6 19}}</ref>
** {{Interstellar Travel}}: Superman is immune to the nature of space and is not inhibited by its lack of gravity. He can travel the reaches of space where no human body could possibly survive without the aid of technology or protection.<ref name=AC_1000/>
** {{Invulnerability}}: Due to the interaction of his dense molecular structure and supercharged bio-electric aura, Superman is nigh-invulnerable to nearly all physical attacks, including bullets, blades, blunt objects, and explosions. Only beings rivaling his own power, Kryptonite-based weaponry, and magic-based attacks can significantly harm Superman. 
*** {{Energy Resistance}}: Superman is highly resistant to most energy-based attacks. On separate occasions he has withstood 300,000 volts and 10 gigawatts of electricity running through him.<ref name=AC2/><ref name ="AC14">{{c|Action Comics Vol 2 14}}</ref> He was even able to deflect Darkseid's full-power [[Omega-Beam]]s off his chest, and was completely unharmed.<ref name=DC-AIS/>
** {{Super-Breath}}: Superman is able to create hurricane force winds by exhaling air from his lungs. He can also chill the air as it leaves his lungs to freeze targets.<ref>{{c|Superman Vol 3 2}}</ref> He can also reverse the process to pull large volumes of air or vapor into his lungs.
** {{Self-Sustenance}}: Superman doesn't need to eat or sleep so long as he has enough solar energy (but is still capable of doing so) and doesn't require oxygen to breathe.<ref>{{c|DC Comics Encyclopedia}}</ref>
** {{Telekinesis|Tactile Telekinesis}}: Superman's body is completely surrounded by an invisible field of electromagnetic energy, protecting him from harm, as well enabling him to perform physical feats that are beyond the mere application force, such as lifting a ship without it  crumbling under its own mass, and retain it's stability. However this can be disrupted by those who can manipulate electromagnetism.<ref name=SMU8/> After White Sun exposure, Superman learned to extend his bio-electric aura to freeze [[Lex Luthor's Warsuit]] with a thought and cause it to break apart with a touch.<ref name=AC-1050/>
** {{Teleportation}}: After White Sun exposure, Superman gained the ability to instantaneously travel across interstellar distances by creating rifts in space-time. When teleported away by Lex Luthor, Superman was able to quickly travel back to Lex though a rift in space.<ref name=AC-1050/>
* '''[[Genesis Fragment|Genesis Enhancement]]:''' When exposed to the powerful, radioactive mineral called [[Genesis Fragment|Genesis]], Superman's powers are temporarily enhanced, allowing him to perform new abilities. Genesis is the crystalized blood of an [[Olgrun (Prime Earth)|Old God]], making it highly mutagenetic. Due to this, Superman cannot absorb large amounts of the energy, unless the mutative radiation is filtered out.<ref name=AC-1054/> 
** {{Energy Construct Creation}}: When supercharged by Genesis, Superman can expand his bio-electric aura to surround himself in a giant energy construct of himself.<ref name=AC-1054/>
** {{Toxic Immunity|Kryptonite Immunity}}: When supercharged by Genesis, Superman can greatly resist the effects of Kryptonite.<ref name=AC-1054/>

{{Navigation
| title = Former Powers
| body  =
* {{Darkness Manipulation}}: For a brief time Kal-El was able to get his hands on the supernatural darkness wielded by [[Wrath II (Prime Earth)|Wrath]] putting it to impressive use while withholding it.<ref>{{c|Action Comics Vol 2 #45}}</ref>
** {{Empathy}}
** {{Matter Absorption}}
*** {{Exorcism}}
** {{Power Distribution|Power Augmentation}}
** {{Telepathy}}
** {{Teleportation}}
*** {{Dimensional Travel}}
* {{Unique Physiology|Altered Physiology}}: Due to [[Vandal Savage (Prime Earth)|Savage]] altering his cellular matrix through foreign radiation, Kal-El was rendered incapable of properly processing yellow sunlight halving his abilities.<ref name=SM-3-43/> Eventually it was discovered that due to the way his genes mutated, he could then be empowered through [[Kryptonite]] to burn off his poisoned biophysical essence and simulate his old powers.
** {{Energy Absorption|Kryptonite Radiation Absorption}}: Superman's Kryptonite charged cells could simulate his old powers but with differences and lethal side effects, using Kryptonite as chemotherapy was killing off his healthy cells as well as the deviant with prolonged exposure.<ref name=SM-3-49/>
*** {{Flight}}
*** {{Invulnerability}}
*** {{Energy Projection|Kryptonite Vision}}
*** {{Enhanced Vision|Modulation Wave Sight}}: He could pick up various transmission frequencies of EM Field.
*** '''Kinetic Field''': Every blow Superman delivered has an explosive concussion effect, shattering objects on impact.
*** {{Superhuman Strength}}
*** {{Superhuman Speed}}
* '''[[Parallax (New Earth)|Parallax Unison]]:''' For a brief amount of time, Clark was fused to the fear entity known as Parallax. With it he had all the power of the yellow entity combined with those of his own, making Superman that much stronger than before.<ref>{{c|Superman Vol 4 29}}&ndash;{{c|Superman Vol 4 30|30}}</ref>
}}

| Abilities               = 
* {{Genius Level Intellect}}: Superman has shown incredible intelligence and computational abilities; his mind works sharply and with extreme speed relative to earth-humans. Superman's analytical powers enable him to read information directly from machines With careful usage of his heat vision, he can even reprogram machines.{{Citation}}
* {{Hand-to-Hand Combat (Advanced)}}: Superman is an excellent hand-to-hand fighter, utilizing a fighting style that directly complements his superpowers. Superman's combat versatility allows him to adapt to almost any foe and beat them. He was trained by Wioska, a former combat instructor of [[Apokolips|Apokolips]],<ref>{{c|Superman: Man of Tomorrow Vol 1 7 (Digital)}}</ref> [[Wonder Woman (Diana Prince)|Wonder Woman]] in Amazonian combat,<ref>{{c|Wonder Woman Vol 4 37}}</ref><ref>{{c|Superman Unchained Vol 1 8}}</ref> and was taught by [[Kryl-Ux (Prime Earth)|Kryl-Ux]] how to fight like a [[Warworld]] gladiator<ref name=AC-1039>{{c|Action Comics Vol 1 1039}}</ref>. He even learned some advanced martial arts techniques from [[Bruce Wayne (Prime Earth)|Batman.]]<ref>{{c|Justice League Vol 2 19}}</ref><ref name=SMU8/><ref>{{c|Batman/Superman: World's Finest Vol 1 19}}</ref> Batman even stated that Clark can take care of himself, even without powers.<ref>{{c|Batman/Superman Annual Vol 1 2}}</ref> In weakened state, Kal got the upper hand on both [[Henry Henshaw (Prime Earth)|Cyborg Superman]] and the [[Eradicator II (Prime Earth)|Eradicator]], while suffering from blindness.<ref>{{c|Action Comics Vol 1 981}}</ref> Superman has shown capable of fighting blind while weakened on multiple occasions.<ref>{{c|Action Comics Vol 1 982}}</ref><ref>{{c|Action Comics Vol 1 983}}</ref> He has fought the likes of [[Teth-Adam (Prime Earth)|Black Adam]] to a stalemate.<ref>{{c|Justice League: Endless Winter Vol 1 2}}</ref>
** {{Martial Arts|Boxing}}: Superman has been trained his father [[Jonathan Kent (Prime Earth)|Johathan Kent]]<ref>{{c|Superman/Wonder Woman Vol 1 13}}</ref> in boxing<ref>{{c|Action Comics Vol 2 16}}</ref>
** {{Martial Arts|Jujitsu}}<ref name=SMU8/>
** {{Martial Arts|Wrestling}}<ref>{{c|Superman Vol 3 47}}</ref>
* {{Indomitable Will}}: Superman has shown to have a powerful spirit, free of corruption and temptation. Having been raised by a kindly Kansas farmer and his wife, he was taught to protect life and help others. He is very optimistic and never gives up, even when things look bad. Superman has learned how to place mental inhibitors on his own power so that he doesn't accidentally "flex and cause the tectonic plates to shift." To him, they are the barriers that allow him to feel human enough to live a healthy life, and he only ever releases one of them when fighting a particularly powerful opponent or performing a difficult task.
* {{Intimidation}}
* {{Investigation}}: Both as a superhero and as a journalist, Superman is an expert at gathering evidence from all kinds of sources.<ref>{{c|Action Comics Vol 2 45}}</ref>
* {{Journalism}}: Clark is one of the best reporters on the ''Daily Planet'', rivaled only by his wife Lois Lane. He gained a {{WP|Pulitzer Prize}} at least in one occasion.
* {{Leadership}}: Superman has proven many times over his ability to command respect and inspire others with his charisma, ardor and idealism. He often serves as the ''primus inter pares'' of the Justice League.
* {{Acrobatics}}<ref>{{c|Batman/Superman Annual Vol 1 2}}</ref><ref>{{c|Action Comics Vol 1 1039}}</ref><ref>{{c|Action Comics Vol 1 1046}}</ref>
* {{Meditation}}<ref>{{c|Justice League Vol 4 23}}</ref><ref>{{c|Action Comics Vol 1 1063}}</ref>
* {{Multilingualism}}: Superman speaks Mandarin Chinese.<ref name=AC983/>
* {{Weaponry}}<ref>{{c|Action Comics Vol 2 23}}</ref><ref>{{c|Superman Unchained Vol 1 7}}</ref>
** {{Stick Fighting}}<ref>{{c|Wonder Woman Vol 4 37}}</ref><ref>{{c|Trinity Vol 2 18}}</ref>
** {{Swordsmanship}}<ref>{{c|Action Comics Vol 1 1040}}</ref><ref>{{c|Action Comics Vol 1 1041}}</ref><ref>{{c|Action Comics Vol 1 1042}}</ref>
| Weaknesses              = {{Kryptonian Physiology|Character=Superman|weak=y}}
* {{Dependency|Solar Energy Dependency}}: Kryptonian abilities will eventually weaken without replenishing their energy reserves with normal (yellow) sun radiation. When exposed to the same red solar radiation as Krypton's red sun Rao, it causes Kryptonians to lose their powers within a large amount of exposure until exposure to yellow sunlight reverses this effect. When exposed to orange solar radiation from a different galaxy, Superman's strength and abilities will drastically diminished, eventually weakening him. The radiation also affects his intellectual capacity and reduces his ability to think.<ref>{{c|Superman Vol 5 27}}</ref>
* {{Power Instability|Sensory Overload}}: Superman's superhumanly acute senses can be overtaxed to the point of pain when encountered by extreme external stimuli such as hypersonics (hearing), balance, and sight (flashbang).
* {{Illness|Radiation Poisoning}}: {{Formerly}} Superman was exposed to deadly radiation during his encounter with the aliens from the Breach, causing him to rapidly age. He slowed the progression of his condition via regular infusions of rejuvanating magic from [[Enchantress (Prime Earth)|The Enchantress]], but this was undone by a sorceress working for Mongul.<ref name=AC_1036/> He was eventually cured by Genesis.<ref>{{c|Superman: Warworld Apocalypse Vol 1 1}}</ref>
** {{Power Limitation}}: {{Formerly}} Superman's sickness caused his powers to fade and weaken. While still significantly stronger than a human, he was less powerful than an average healthy Kryptonian until his radiation poisoning was cured.<ref>{{c|Superman and the Authority Vol 1 2}}</ref>
| Other                   = * {{Mental Disorder|Emotional Instability}}: Superman was exposed to synthetic [[Red Kryptonite]], which unlike genuine Red Kryptonite did not wear off after twenty-four hours. As a result of the Kryptonite, Superman can potentially fly into a violent rage if he becomes overstressed. In this state he loses all self-control and could kill someone.<ref>{{c|Superman Vol 6 12}}</ref><ref>{{c|Superman Vol 6 20}}</ref><ref>{{c|Superman Vol 6 21}}</ref>

| Equipment               = 
| Transportation          = 
| Weapons                 = 
* '''[[Yellow Lantern Ring]]''' {{Formerly}}: While a part of the Sinestro Corps, Superman earned a ring for his ability to inspire great fear in others.<ref>{{c|Sinestro Vol 1 18}}</ref>

| Notes                   = 
{{PEBoilerplate
 | Page        = Kal-L (Earth-Two)
 | Name        = Superman
 | Creator1    = Jerry Siegel
 | Creator2    = Joe Shuster
 | First       = Action Comics Vol 1 1
 | NewIssue    = Flashpoint Vol 2 5
 | Creator4    = Geoff Johns
 | Creator5    = Andy Kubert
}}
* [[Jonathan Osterman (Watchmen)|Doctor Manhattan]] discovered that Superman is a pivotal constant in the universe. Despite changes to timeline, the [[Metaverse]] consistently preserves Superman’s existence and core history by expanding itself, creating new realities to preserve every era of Superman, emphasizing his role as a source of hope that anchors reality and inspiring heroism across all realities.<ref name="DC12"/>
| Trivia                  =  
* Kal-El literally translates to "''Star Child''" from the [[Kryptonese]].
* Clark considers [[Batman (Bruce Wayne)|Bruce Wayne]] to be his best friend.<ref name=BMSM3/>
* [[Batman (Bruce Wayne)|Batman]] considers Superman to be his favorite superhero.<ref>{{c|Detective Comics Vol 1 1017}}</ref>
* Superman has been an active crime-fighter for at least two decades.<ref>{{C|Superman and the Authority Vol 1 1}}</ref>
* Superman believes [[Bruce Wayne (Prime Earth)|Batman]] is a better man than him.<ref>{{c|Batman Vol 3 36}}</ref>
* Clark is a fan of [[wikipedia:Clint Eastwood|Clint Eastwood]].<ref>{{c|Adventures of the Super Sons}}</ref>
* Clark considers [[Richard Grayson (Prime Earth)|Dick Grayson]] to be his favorite Robin.<ref>{{c|Superman: Up in the Sky Vol 1 6}}</ref>
* Superman has been credited for being a highly skilled fighter/warrior, by impressing the likes of [[Adam Blake (Prime Earth)|Captain Comet]],<ref>{{c|Action Comics Vol 2 11}}</ref> [[Batman (Bruce Wayne)|Batman]],<ref>{{c|Batman/Superman Annual Vol 1 2}}</ref><ref>{{c|Trinity Vol 2 17}}</ref> [[Wonder Woman (Diana Prince)|Wonder Woman]],<ref>{{c|Trinity Vol 2 17}}</ref> [[Artemis, Goddess of the Hunt (Prime Earth)|Moon]],<ref>{{c|Superman/Wonder Woman Vol 1 25}}</ref> [[Pax Galactica]],<ref>{{c|Action Comics Vol 2 23}}</ref> [[Dru-Zod (Prime Earth)|General Zod]],<ref>{{c|Action Comics Vol 1 983}}</ref> '''Grikus the Undefeated,'''<ref>{{c|Superman: Man of Tomorrow Vol 1 11 (Digital)}}</ref> '''the Tournament Judge of Warworld,'''<ref>{{c|Batman/Superman Annual Vol 1 1}}</ref> '''Wioska,''' <ref>{{c|Superman: Man of Tomorrow Vol 1 7 (Digital)}}</ref> and [[Desaad (Prime Earth)|Desaad]].<ref>{{c|Action Comics Vol 1 1049}}</ref>
* [[Khalid Nassour (Prime Earth)|Khalid Nassour]] served as Superman's therapist for a time after outing his secret identity.<ref>{{c|Superman Vol 5 26}}</ref><ref>{{c|Superman Vol 5 24}}</ref>
* Every sunday in Smalville, Clark and his Father used to sit on the couch and watch a classic Western together.<ref>{{c|Superman Vol 6 10}}</ref>
| Recommended = * {{v|Superman Vol 3}}
* {{v|Action Comics Vol 2}}
* {{v|Justice League Vol 2}}
* {{v|Superman Vol 4}}
* {{v|Action Comics Vol 1}}
* {{v|Justice League Vol 3}}
* {{v|Justice League Vol 4}}
* {{v|Superman Vol 5}}
* {{v|Batman/Superman: World's Finest Vol 1}}
* {{c|Superman: Kal-El Returns Special Vol 1 1}}
* {{v|Superman Vol 6}}
| Links = }}
{{Superman Family Member}}
{{Daily Planet Staff members}}
{{Justice League Member}}
{{Authority members}}
{{Sinestro Corps Member}}{{Legion of Super-Heroes Member}}
{{Black Lantern Corps Member}}
[[pt-br:Kal-El (Terra Principal)]]
[[uk:Кел-Ел (Прайм Земля)]]
[[Category:New 52 Characters]]
[[Category:Diana of Paradise Island's Love Interests]]
[[Category:Formerly Deceased]]
[[Category:Farmers]]
[[Category:Firefighters]]
[[Category:Helmet of Nabu Hosts]]
[[Category:Parallax Hosts]]
[[Category:Reporters]]
[[Category:Gestalt Characters]]
[[Category:Lois Lane's Love Interests]]
[[Category:Smallville High School students]]
[[Category:Sanctuary patients]]
[[Category:New Gods]]


"""
```

```python
# Função para remover tags e manter apenas o texto
def extract_text(raw_text):
    # Remove comentários (<!-- ... -->)
    no_comments = re.sub(r'<!--.*?-->', '', raw_text, flags=re.DOTALL)
    # Remove tags de template e referência ({{ ... }})
    no_templates = re.sub(r'{{.*?}}', '', no_comments, flags=re.DOTALL)
    # Remove tags de formatação ([[...]] e <...>)
    no_format_tags = re.sub(r'\[\[.*?\||\[\[|\]\]', '', no_templates)
    no_format_tags = re.sub(r'<.*?>', '', no_format_tags)
    # Substitui múltiplas quebras de linha por uma única
    cleaned_text = re.sub(r'\n+', '\n', no_format_tags).strip()
    return cleaned_text

# Extrai o texto
cleaned_text = extract_text(code)
print(cleaned_text)
```
